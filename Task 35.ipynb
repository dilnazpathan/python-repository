```python
# 9. Remove duplicate customer records while keeping the first occurrence.

df = df.drop_duplicates(subset = "customer_id",keep ="first")
df["customer_id"].duplicated().sum()
```

```python
# 10.Find all unique values in the customer_segment column and identify inconsistent categories such as Consumer, consumer, consumer , etc.

df["customer_segment"].unique()
```

```python
df["customer_segment"].astype(str).str.strip().str.lower().value_counts()
```

```python
# 11.Clean the customer_segment column by standardizing capitalization and whitespace.

df["customer_segment"]=df["customer_segment"].str.strip().str.title()
df["customer_segment"]
```

```python
# 12.Find all unique values in the gender column and convert them into consistent categories such as Male and Female.

df["gender"].unique()
```

```python
df["gender"] = df["gender"].str.strip().str.title()
df["gender"].unique()
```

```python
# 13.Clean the is_active column so values such as Yes, yes, Y, 1 become 1, and No, no, N, 0 become 0.

df["is_active"] = (df["is_active"].astype(str).str.strip().str.lower().map({"yes":1,"y":1,"1":1,"no":0,"n":0,"0":0}))
df["is_active"].unique()
```

```python
df["is_active"].value_counts()
```

```python
# 14.Convert the churn_flag column from categorical values (Yes/No) into a binary numerical column (1/0).

df["churn_flag"] = (df["churn_flag"].astype(str).str.strip().str.lower().map({"Yes":1,"No":0}))
df["churn_flag"].unique()
```

```python
#15.Identify all values in income_usd that cannot be directly interpreted as numbers.

temp = pd.to_numeric(df["income_usd"],errors="coerce")
df[df["income_usd"].notna() & temp.isna()]["income_usd"].unique()
```

```python
# 16.Clean income_usd by removing currency symbols, commas, and invalid text, then convert it to a numeric column using pd.to_numeric().

df["income_usd"] = df["income_usd"].astype(str).str.replace(r"[$,]","",regex = True).str.replace(r"[^\d.-]","",regex=True)
df["income_usd"] = pd.to_numeric(df["income_usd"],errors="coerce")
df["income_usd"].head()
```

```python
# 17.Clean the Age column so values such as "35 years" become numeric values.

df["age"] = pd.to_numeric(df["age"].astype(str).str.replace("years","").str.strip(),errors="coerce")
df["age"]
```

```python
# 18.Identify unrealistic age values using NumPy/Pandas and replace them with NaN.

import numpy as np
df["age"] = df["age"].where(df["age"].between(18,100),np.nan)
df["age"]
```

```python
# 19.Clean the credit_score column by removing text such as "pts" and converting the column to numeric.

df["credit_score"] = pd.to_numeric(df["credit_score"].astype(str).str.replace("pts", "").str.strip(),errors="coerce")
df["credit_score"]
```

```python
#20.Detect outliers in income_usd using the IQR method with NumPy/Pandas.

import numpy as np
Q1 = df["income_usd"].quantile(0.25)
Q3 = df["income_usd"].quantile(0.75)
IQR = Q3-Q1
lower = Q1 - 1.5 * IQR
upper = Q3 - 1.5 * IQR
outlier = df[(df["income_usd"] < lower) | (df["income_usd"] > upper)]
outlier
```

```python
# 21.Detect outliers in credit_score using the IQR method.

Q1 = df["credit_score"].quantile(0.25)
Q3 = df["credit_score"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 - 1.5 * IQR
outlier = df[(df["credit_score"]<lower)|(df["credit_score"]>upper)]
outlier
```

```python
# 22.Detect suspicious values such as -999 and 99999 in tenure_years and satisfaction.

suspicious_values = [-999,99999]
print("suspicious tenure_years:")
print(df[df["tenure_years"].isin(suspicious_values)])
print("suspicious satisfaction:")
print(df[df["satisfaction"].isin(suspicious_values)])
```

```python
# 23.Replace invalid/suspicious numerical values with NaN.

import numpy as np
df["tenure_years"] = df["tenure_years"].replace([-999,99999],np.nan)
df["satisfaction"] = df["satisfaction"].replace([-999,99999],np.nan)
df[["tenure_years","satisfaction"]]
```

```python
# 24.Fill missing numerical values using the median of each numerical column.

num_cols = df.select_dtypes(include ="number").columns
df[num_cols] = df[num_cols].fillna(df[num_cols].median())
df[num_cols].isna().sum()
```

```python
# 25.Fill missing categorical values using the mode of each categorical column.

cat_cols = df.select_dtypes(include = "object").columns
for col in  cat_cols:
    df[col].fillna(df[col].mode()[0])
df[cat_cols].isna().sum()
```

```python
# 26.Clean the signup_date column, which contains multiple date formats and invalid dates, and convert it to Pandas datetime.

df["signup_date"] = pd.to_datetime(df["signup_date"],errors="coerce")
df["signup_date"].head()
```

```python
# 27.Find rows where signup_date could not be converted into a valid date.

invalid_dates = df[df["signup_date"].isna()]
invalid_dates
```

```python
# 28.Clean the lead_source column by standardizing values such as Website, website, WEB, Referral, referral , Google Ads, and google_ads.

df["lead_source"] = df["lead_source"].str.lower().str.strip()
df["lead_source"] = df["lead_source"].replace({"web":"website","website":"website","referral":"referral","google_ads":"google ads","google ads":"google ads"})
df["lead_source"].value_counts()
```

```python
# 29.Create a final cleaned DataFrame containing:
#• cleaned column names
#• no duplicate IDs
#• standardized categorical values
#• numeric columns converted to numeric types
#• invalid values converted to NaN
#• missing values handled
#• detected outliers handled

df.columns = df.columns.str.lower().str.strip().str.replace(" ","_")
df = df.drop_duplicates(subset="customer_id")
num_cols = df.select_dtypes(include="number").columns
for col in num_cols:
    df[col] = pd.to_numeric(df[col],errors="coerce")
df["tenure_years"] = df["tenure_years"].replace([-999,99999],np.nan)
df["satisfaction"] = df["satisfaction"].replace([-999,99999],np.nan)
cat_cols = df.select_dtypes(include ="object").columns
for col in cat_cols:
    df[col] = df[col].str.lower().str.strip()
num_cols = df.select_dtypes(include="number").columns
for col in num_cols:
    df[col] = df[col].fillna(df[col].median())
cat_cols = df.select_dtypes(include ="object").columns
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
final_df = df.copy()
final_df.head()
```
