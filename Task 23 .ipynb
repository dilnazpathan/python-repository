### Hybrid Inheritance Questions (1-10)

```python
# 1. Create a hybrid inheritance structure with classes A → B and C → D (where D inherits from both B and C).
#Concept
#Hybrid Inheritance (Combination of Single + Multiple Inheritance)
#What to do
#• Create parent class A.
#• Create class B inheriting from A.
#• Create another class C.
#• Create class D inheriting from both B and C.
#How to solve
#1. Create class A.
#2. Add one method.
#3. Create class B(A).
#4. Create class C.
#5. Create class D(B, C).
#6. Create an object of D.
#7. Call methods from all parent classes.

class A:
    def method_A(self):
        print("this is method from class A")
class B(A):
    def method_B(self):
        print("this is method from class B")
class C:
    def method_C(self):
        print("this is method from class C")
class D(B,C):
    def method_D(self):
        print("This is method from class D")
obj = D()
obj.method_A()
obj.method_B()
obj.method_C()
obj.method_D()
```

```python
# 2. Create Person, Employee, Student, and TeachingAssistant using hybrid inheritance.
#Concept
#Real-world Hybrid Inheritance
#What to do
#• Person is the base class.
#• Employee inherits Person.
#• Student inherits Person.
#• TeachingAssistant inherits Employee and Student.
#How to solve
#Hierarchy
#Person
#/ \
#Employee Student
#\ /
#TeachingAssistant
#Steps
#1. Store name in Person.
#2. Employee stores salary.
#3. Student stores marks.
#4. TeachingAssistant stores department.
#5. Display all information.

class Person:
    def __init__(self,name):
        self.name = name
class Employee(Person):
    def __init__(self,name,salary):
        Person.__init__(self,name)
        self.salary = salary
class Student(Person):
    def __init__(self,name,marks):
        Person.__init__(self,name)
        self.marks = marks
class TeachingAssistant(Employee,Student):
    def __init__(self,name,salary,marks,department):
        Person.__init__(self,name)
        self.marks = marks
        self.salary = salary
        self.department = department
    def display(self):
        print("name :", self.name)
        print("salary:", self.salary)
        print("marks:",self.marks)
        print("department:",self.department)
t = TeachingAssistant("dilnaz",50000,90,"IT")
t.display()
        
```

```python
# 3. Create Vehicle → Car, Electric → ElectricCar (ElectricCar inherits from both).
#Concept
#Hybrid Inheritance
#What to do
#• Vehicle is parent.
#• Car inherits Vehicle.
#• Electric is another parent.
#• ElectricCar inherits Car and Electric.
#How to solve
#Hierarchy
#Vehicle
#|
#Car Electric
#\ /
#ElectricCar
#Steps
#1. Vehicle has start().
#2. Car has drive().
#3. Electric has battery().
#4. ElectricCar accesses all methods.

class Vehicle:
    def start(self):
        print("vehicle is starting")
class Car(Vehicle):
    def drive(self):
        print("car is being driven")
class Electric(Vehicle):
    def battery(self):
        print("battery is fully charged")
class ElectricCar(Car,Electric):
    def details(self):
        print("This is an electric car")
e = ElectricCar()
e.start()
e.drive()
e.battery()
e.details()


    
```

```python
# 4. Demonstrate Hybrid Inheritance and print Method Resolution Order (MRO).
#Concept
#Method Resolution Order (MRO)
#What to do
#Create
#A
#|
#B

#C

#D(B,C)
#Print
#D.mro()
#How to solve
#1. Create four classes.
#2. Give same method name in multiple classes.
#3. Print ClassName.mro().
#4. Observe Python's searching order.


class A:
    def show(self):
        print("Method from Class A")
class B(A):
    def show(self):
        print("Method from Class B")
class C:
    def show(self):
        print("Method from Class C")
class D(B, C):
    pass
obj = D()
obj.show()
print("\nMethod Resolution Order (MRO):")
print(D.mro())
```

```python
# 5. Create class A, B, C, D and solve the Diamond Problem using super().
#Concept
#Diamond Problem
#Hierarchy
#A
#/ \
#B C
#\ /
#D
#What to do
#Use super() everywhere.
#How to solve
#1. Create constructor in A.
#2. Call super() in B.
#3. Call super() in C.
#4. Call super() in D.
#5. Create object of D.
#6. Observe A constructor runs only once.


class A:
    def __init__(self):
        print("Constructor of Class A")
class B(A):
    def __init__(self):
        super().__init__()
        print("Constructor of Class B")
class C(A):
    def __init__(self):
        super().__init__()
        print("Constructor of Class C")
class D(B, C):
    def __init__(self):
        super().__init__()
        print("Constructor of Class D")
obj = D()
print("\nMethod Resolution Order (MRO):")
print(D.mro())
```

```python
# 6. Create a banking system using hybrid inheritance.
#Concept
#Hybrid + Real-world Example
#What to do
#Account
#|
#Savings
#Loan
#SavingsLoanAccount
#How to solve
#1. Account stores account number.
#2. Savings stores balance.
#3. Loan stores loan amount.
#4. SavingsLoanAccount inherits both.
#5. Print all details.

class Account:
    def __init__(self,account_number):
        self.account_number = account_number
class Savings (Account):
    def __init__(self,account_number,balance):
        Account.__init__(self,account_number)
        self.balance = balance
class Loan:
    def __init__(self,loan_amount):
        self.loan_amount = loan_amount
class SavingsLoanAccount(Savings,Loan):
    def __init__(self,account_number,balance,loan_amount):
        Savings.__init__(self,account_number,balance)
        Loan.__init__(self,loan_amount)
    def display(self):
        print("account number:",self.account_number)
        print("balance:",self.balance)
        print("loan amount:", self.loan_amount)
obj = SavingsLoanAccount(123456,50000,200000)
obj.display()
        
    
```

```python
# 7. Create Shape, Color, Circle, and ColoredCircle.
#Concept
#Hybrid Inheritance
#What to do
#Shape
#|
#Circle
#color
#ColoredCircle
#How to solve
#1. Shape has area().
#2. Color has color().
#3. Circle inherits Shape.
#4. ColoredCircle inherits Circle and Color.
#5. Display area and color.

class Shape:
    def area(self,radius):
        area = 3.14 * radius * radius
        print("area of circle:",area)
class Circle(Shape):
    def draw(self):
        print("circle is drawn")
class Color:
    def color(self,color_name):
        print("color of circle:",color_name)
class ColoredCircle(Circle,Color):
    def display(self):
        print("this is colored circle")
obj = ColoredCircle()
obj.area(7)
obj.draw()
obj.color("blue")
obj.display()

    
        
```

```python
# 8. Create constructors at every level.
#Concept
#Constructor Chaining
#What to do
#Every class has its own constructor.
#How to solve
#1. Create constructor in every class.
#2. Use super() everywhere.
#3. Print constructor execution order.
#4. Observe execution using MRO.


class A:
    def __init__(self):
        super().__init__()
        print("Constructor of Class A")
class B(A):
    def __init__(self):
        super().__init__()
        print("Constructor of Class B")
class C(A):
    def __init__(self):
        super().__init__()
        print("Constructor of Class C")
class D(B, C):
    def __init__(self):
        super().__init__()
        print("Constructor of Class D")
obj = D()
print("\n Method Resolution Order (MRO):")
print(D.mro())
```

```python
# 9. Override methods in intermediate classes.
#Concept
#Method Overriding
#What to do
#A
#|
#B

#C
#D(B,C)
#All classes have
#show()
#How to solve
#1. Create same method.
#2. Override in B.
#3. Override in C.
#4. Call from D.
#5. Observe MRO.


class A:
    def show(self):
        print("Show method from Class A")
class B(A):
    def show(self):
        print("Show method from Class B")
class C:
    def show(self):
        print("Show method from Class C")
class D(B, C):
    pass
obj = D()
obj.show()
print("\nMethod Resolution Order (MRO):")
print(D.mro())
```

```python
# 10. Create a Device → Phone, Camera → SmartPhone hierarchy.
#Concept
#Hybrid Inheritance
#What to do
#Device
#|
#Phone

#Camera

#SmartPhone
#How to solve
#1. Device has power_on().
#2. Phone has call().
#3. Camera has click().
#4. SmartPhone accesses all methods.

class Device:
    def power_on(self):
        print("Device is powered on.")
class Phone(Device):
    def call(self):
        print("Making a phone call.")
class Camera:
    def click(self):
        print("Photo clicked.")
class SmartPhone(Phone, Camera):
    def features(self):
        print("SmartPhone has calling and camera features.")
obj = SmartPhone()
obj.power_on()    
obj.call()        
obj.click()       
obj.features()    
```

### Duck Typing Questions (11-20)

```python
# 11. What is Duck Typing? Explain with a simple example.
#Concept
#Duck Typing
#"If an object behaves like a duck, Python treats it like a duck."
#What to do
#Create two unrelated classes having the same method.
#How to solve
#1. Create Dog.
#2. Create Robot.
#3. Both have speak().
#4. Write one function calling obj.speak().
#5. Pass both objects.

class Dog:
    def speak(self):
        print("dog says:wooofff!")
class Robot:
    def speak(self):
        print("robot says: beep boop!")
def make_speak(obj):
    obj.speak()
dog = Dog()
robot = Robot()
make_speak(dog)
make_speak(robot)
```

```python
# 12. Create Dog and Robot both having speak().
#Concept
#Duck Typing
#What to do
#Common method
#speak()
#How to solve
#1. Dog → speak().
#2. Robot → speak().
#3. Function accepts any object.
#4. Call obj.speak().


class Dog:
    def speak(self):
        print("Dog says: Woof!")
class Robot:
    def speak(self):
        print("Robot says: Beep Boop!")
def make_speak(obj):
    obj.speak()
dog = Dog()
robot = Robot()
make_speak(dog)
make_speak(robot)
```

```python
# 13. Create Car and Bicycle both having start().
#Concept
#Duck Typing
#What to do
#Common method
#start()
#How to solve
#1. Car defines start().
#2. Bicycle defines start().
#3. Function starts any vehicle.

class Car:
    def start(self):
        print("car started")
class Bicycle:
    def start(self):
        print("bicycle started")
def start_vehicle(vehicle):
    vehicle.start()
car = Car()
bicycle = Bicycle()
start_vehicle(bicycle)
start_vehicle(car)
```

```python
# 14. Create PDFFile and WordFile both having read().
#Concept
#Duck Typing
#What to do
#One function
#open_file(file)
#How to solve
#1. PDFFile → read().
#2. WordFile → read().
#3. Function calls read().

class PDFFile:
    def read(self):
        print("reading PDF file")
class WordFile:
    def read(self):
        print("reading word file")
def open_file(file):
    file.read()
p = PDFFile()
w  = WordFile()
open_file(p)
open_file(w)
```

```python
# 15. Create Circle and Rectangle both having area().
#Concept
#Duck Typing
#What to do
#Function
#calculate(obj)
#How to solve
#1. Circle implements area().
#2. Rectangle implements area().
#3. Function calls area().
#4. Print returned value.


class Circle:
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return 3.14 * self.radius * self.radius
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width
    def area(self):
        return self.length * self.width
def calculate(obj):
    print("Area =", obj.area())
circle = Circle(7)
rectangle = Rectangle(10, 5)
calculate(circle)
calculate(rectangle)
```

```python
# 16. Create CreditCard and UPI both having pay().
#Concept
#Duck Typing
#What to do
#Payment system
#How to solve
#1. CreditCard → pay().
#2. UPI → pay().
#3. Function processes payment using pay().


class CreditCard:
    def pay(self):
        print("Payment made using Credit Card.")
class UPI:
    def pay(self):
        print("Payment made using UPI.")
def process_payment(method):
    method.pay()
credit = CreditCard()
upi = UPI()
process_payment(credit)
process_payment(upi)
```

```python
# 18. Create Logger classes using log(message).
#Concept
#Duck Typing
#What to do
#Two unrelated logger classes.
#How to solve
#1. FileLogger → log().
#2. ConsoleLogger → log().
#3. Function accepts logger object.
#4. Calls log(message).


class FileLogger:
    def log(self, message):
        print("File Logger:", message)
class ConsoleLogger:
    def log(self, message):
        print("Console Logger:", message)
def write_log(logger, message):
    logger.log(message)
file_logger = FileLogger()
console_logger = ConsoleLogger()
write_log(file_logger, "User logged in")
write_log(console_logger, "User logged in")
```

```python
# 19. Create a hybrid inheritance structure and apply Duck Typing.
#Concept
#Hybrid + Duck Typing
#What to do
#Create hybrid inheritance.
#Create another unrelated class with same method.
#How to solve
#1. Create hybrid hierarchy.
#2. Final child has method show().
#3. Another unrelated class also has show().
#4. One function calls show() for both.


class A:
    pass
class B(A):
    pass
class C:
    pass
class D(B, C):
    def show(self):
        print("Show method from Hybrid Class D")
class AnotherClass:
    def show(self):
        print("Show method from AnotherClass")
def display(obj):
    obj.show()
d = D()
another = AnotherClass()
display(d)
display(another)
```

```python
# 20. Create Employee hierarchy using Hybrid Inheritance and use Duck Typing.
#Concept
#Hybrid Inheritance + Duck Typing
#What to do
#Hierarchy
#Person
#|
#Employee

#Manager

#HR

#TeamLead
#Another class
#Freelancer
#Both TeamLead and Freelancer have
#work()
#How to solve
#1. Build hybrid inheritance.
#2. Final child defines work().
#3. Freelancer also defines work().
#4. Write a function:
#def do_work(obj):
#obj.work()
#5. Pass both objects to the function.
#6. Observe that the same function works for unrelated classes because of Duck Typing.


class Person:
    pass
class Employee(Person):
    pass
class Manager:
    pass
class HR(Employee, Manager):
    pass
class TeamLead(HR):
    def work(self):
        print("TeamLead is managing the team.")
class Freelancer:
    def work(self):
        print("Freelancer is working on a project.")
def do_work(obj):
    obj.work()
team_lead = TeamLead()
freelancer = Freelancer()
do_work(team_lead)
do_work(freelancer)
```
