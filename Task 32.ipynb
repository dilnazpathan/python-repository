### Part 2 — Detect & Fix Data Quality Problems

```python
# Q6. Duplicate investigation
#Find how many duplicate rows exist.
#Do not remove them yet.
#Use: duplicated()

df.duplicated().sum()
```

```python
# Q7. Remove duplicates
#Remove all duplicate records and store the cleaned result back into df.
#Then verify that duplicates no longer exist.
#Use: drop_duplicates(), duplicated()

df = df.drop_duplicates()
df.duplicated().sum()
```

```python
# Q8. Numeric column investigation
#Select only numeric columns.
#Identify which columns are currently numeric and which potentially numeric columns have
#become object because of invalid values.
#Use: select_dtypes()

numeric_columns = df.select_dtypes(include="number")
numeric_columns.columns
```

```python
object_columns = df.select_dtypes(include="object")
object_columns.columns
```

```python
# Q9. Fix Age
#The Age column contains:
#• Valid numbers
#• Missing values
#• Invalid text such as unknown, N/A, etc.
#Convert Age into a proper numeric column.
#Invalid values should become missing values.
#Then fill missing Age values using the median age.
#Use: pd.to_numeric(), median(), fillna()

df["Age"] = pd.to_numeric(df["Age"],errors="coerce")
df["Age"].isna().sum()
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Age"].isna().sum()
```

```python
# Q10. Fix Annual Income
#Clean Annual_Income.
#Requirements:
#1. Convert invalid values to missing.
#2. Convert the column to a numeric datatype.
#3. Calculate the median.
#4. Fill missing values using the median.
#Use: pd.to_numeric(), median(), fillna(), astype()

df["Annual_Income"] = pd.to_numeric(df["Annual_Income"],errors="coerce")
df["Annual_Income"].isna().sum()
```

```python
median_income = df["Annual_Income"].median()
df["Annual_Income"].dtype
```

```python
df["Annual_Income"] = df["Annual_Income"].fillna(median_income)
```

```python
df["Annual_Income"] = df["Annual_Income"].astype(float)
df[["Annual_Income"]].head(10)
```

### Part 3 — Numerical Cleaning

```python
# Q11. Clean Account Balance
#The Account_Balance column contains invalid strings and missing values.
#Clean the column and replace missing values with the mean Account Balance.
#Use:
#• pd.to_numeric()
#• mean()
#• fillna()

df["Account_Balance"] = pd.to_numeric(df["Account_Balance"],errors="coerce")
df["Account_Balance"] = df["Account_Balance"].fillna(df["Account_Balance"].mean())
df["Account_Balance"].isna().sum()
```

```python
# Q12. Clean Credit Score
#Clean Credit_Score.
#Invalid values should become missing.
#Then replace missing values with the median Credit Score.
#Finally convert the column to integer.
#Use:
#• pd.to_numeric()
#• median()
#• fillna()
#• astype()

df["Credit_Score"] = pd.to_numeric(df["Credit_Score"],errors = "coerce")
df["Credit_Score"] = df["Credit_Score"].fillna(df["Credit_Score"].median())
df["Credit_Score"] = df["Credit_Score"].astype(int)
df["Credit_Score"].isna().sum(),df["Credit_Score"].dtype
```

```python
# Q13. Clean Monthly Spend
#Clean Monthly_Spend.
#Then:
#• Convert it to numeric.
#• Replace invalid values.
#• Replace missing values using the median.
#• Make sure the final column is numeric.

df["Monthly_Spend"] = pd.to_numeric(df["Monthly_Spend"],errors="coerce")
df["Monthly_Spend"] = df["Monthly_Spend"].fillna(df["Monthly_Spend"].median())
df["Monthly_Spend"].isna().sum(),df["Monthly_Spend"].dtype
```

```python
# Q14. Clean Transactions
#The Transactions column contains invalid strings.
#Convert it to numeric and replace invalid values with the median number of transactions.
#Then convert the final column to integer.

df["Transactions"] = pd.to_numeric(df["Transactions"],errors="coerce")
df["Transactions"] = df["Transactions"].fillna(df["Transactions"].median())
df["Transactions"] = df["Transactions"].astype(int)
df["Transactions"].isna().sum(),df["Transactions"].dtype
```

```python
# Q15. Clean Satisfaction Score
#Clean Satisfaction_Score.
#Replace missing/invalid values using the mean Satisfaction Score.
#Then convert the column to float.

df["Satisfaction_Score"] = pd.to_numeric(df["Satisfaction_Score"],errors="coerce")
df["Satisfaction_Score"] = df["Satisfaction_Score"].fillna(df["Satisfaction_Score"].mean())
df["Satisfaction_Score"] = df["Satisfaction_Score"].astype(float)
df["Satisfaction_Score"].isna().sum(),df["Satisfaction_Score"].dtype
```

### Part 4 — Date Cleaning

```python
# Q16. Investigate Join_Date
#Inspect the Join_Date column.
#Identify the problems caused by:
#• Missing dates
#• Different date formats
#• Invalid date strings
#Do not clean it yet.

df["Join_Date"].head(20),df["Join_Date"].isna().sum(),df["Join_Date"].dtype
df["Join_Date"].value_counts().head(10)
```

```python
# Q17. Convert Join_Date
#Convert Join_Date into a proper Pandas datetime column.
#Any invalid dates should become missing values.
#Use: pd.to_datetime()

df = pd.read_csv("C:\\Users\\DELL\\Downloads\\customer_analytics - customer_analytics.csv")
df["Join_Date"] = pd.to_datetime(df["Join_Date"], errors="coerce")
df["Join_Date"].isna().sum(),df["Join_Date"].dtype
```

```python
# Q18. Handle missing dates
#After conversion, find how many missing dates remain.
#Then replace missing dates appropriately using a reasonable date-based strategy.

df["Join_Date"] = df["Join_Date"].fillna(df["Join_Date"].median())
df["Join_Date"].isna().sum()
```

### Part 5 — Categorical Cleaning

```python
# Q19. Clean City
#The City column contains values such as:
#Mumbai
#mumbai
#Mumbai
#mumbai
#Standardize the values so that logically identical cities are represented consistently.
#Then check the unique frequency distribution.
#Use: replace(), value_counts()

df["City"] = df["City"].replace({"mumbai":"Mumbai","hyderabad":"Hyderabad","bangalore":"Bangalore","pune":"Pune","delhi":"Delhi","kolkata":"Kolkata","chennai":"Chennai"})
df["City"].value_counts(dropna=False)
```

```python
# Q20. Analyze Segment
#Check the frequency of:
#Retail
#SME
#Corporate
#Premium
#Also determine whether missing values exist.
#Then replace missing Segment values with the most appropriate category based on your
#analysis.
#Use: value_counts(), fillna()

df["Segment"] = df["Segment"].fillna("Retail")
df["Segment"].isna().sum(),df["Segment"].value_counts()
```

```python
#Q21. Clean Active
#Active is supposed to be a binary column, but it contains:
#Yes
#No
#Y
#N
#YES
#NO
#yes
#no
#1
#0
#Transform these into a consistent representation:
#Yes
#No
#Then verify the result with value_counts().
#Use: replace(), value_counts()

df["Active"] = df["Active"].replace({"Yes":"Yes","YES":"Yes","yes":"Yes","Y":"Yes",1:"Yes","1":"Yes","No":"No","NO":"No","no":"No",
                                     "N":"No",0:"No","0":"No"})
df["Active"].value_counts(dropna=False)
                                     
```

```python
#Q22. Clean Risk_Level
#Check the distribution of:
#High
#Medium
#Low
#Verify whether missing values exist.
#Then display the final class distribution.

df["Risk_Level"].value_counts(dropna=False)
df["Risk_Level"].isna().sum()
```

### Part 7 — Create New Columns

```python
# Q26. Create Annual Income in Lakhs
#Create:
#Income_Lakhs
#using:
#AnnualIncome / 100000
#Then check the first 10 records.

df["Customer_AnnualIncome"] = pd.to_numeric(df["Customer_AnnualIncome"],errors="coerce")
df["Income_Lakhs"] = df["Customer_AnnualIncome"] / 100000
df[["Customer_AnnualIncome","Income_Lakhs"]].head(10)
```

```python
# Q27. Create Monthly Income
#Create:
#MonthlyIncome
#using:
#AnnualIncome / 12
#Then create:
#Income_After_Spending
#using:
#MonthlyIncome - MonthlySpend

df["Customer_MonthlySpend"] = pd.to_numeric(df["Customer_MonthlySpend"], errors="coerce")
df["MonthlyIncome"] = df["Customer_AnnualIncome"] / 12
df["Income_After_Spending"] = df["MonthlyIncome"] - df["Customer_MonthlySpend"]
df[["MonthlyIncome","Customer_MonthlySpend","Income_After_Spending"]].head(10)
```

```python
# Q28. Create Balance-to-Income Ratio
#Create:
#Balance_Income_Ratio
#using:
#AccountBalance / AnnualIncome
#Then inspect the first 10 rows.
#Think carefully about division by zero and missing values.

df["Customer_AccountBalance"] = pd.to_numeric(df["Customer_AccountBalance"], errors="coerce")
df["Customer_AnnualIncome"] = pd.to_numeric(df["Customer_AnnualIncome"], errors="coerce")
df["Balance_Income_Ratio"] = df["Customer_AccountBalance"] / df["Customer_AnnualIncome"].replace(0,float("nan"))
df[["Customer_AccountBalance", "Customer_AnnualIncome", "Balance_Income_Ratio"]].head(10)
```

```python
# Q30 — Ultimate Pandas Cleaning Project
#You are a Data Scientist at a bank. Your raw customer dataset contains 2,000+ records with:
#• Missing values
#• Duplicate records
#• Invalid numerical values
#• Mixed datatypes
#• Invalid dates
#• Mixed date formats
#• Inconsistent categorical values
#• Binary text inconsistencies
#• Unnecessary identifier columns

df = pd.read_csv("C:\\Users\\DELL\\Downloads\\customer_analytics - customer_analytics.csv")
df = df.drop_duplicates()
numeric_cols = ["Age","Annual_Income","Account_Balance","Credit_Score","Tenure_Years","Monthly_Spend","Transactions",
    "Satisfaction_Score"]
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")
for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())
df["Join_Date"] = pd.to_datetime(df["Join_Date"],errors="coerce")
df["Join_Date"] = df["Join_Date"].fillna(df["Join_Date"].median())
df["City"] = df["City"].replace({"mumbai": "Mumbai","hyderabad": "Hyderabad","bangalore": "Bangalore","pune": "Pune",
    "delhi": "Delhi","kolkata": "Kolkata","chennai": "Chennai"})
df["Segment"] = df["Segment"].fillna(df["Segment"].mode()[0])
df["Active"] = df["Active"].replace({"Yes": "Yes","YES": "Yes","yes": "Yes","Y": "Yes","1": "Yes",
    1: "Yes","No": "No","NO": "No","no": "No","N": "No","0": "No",0: "No"})
df = df.rename(columns={"Annual_Income": "AnnualIncome","Account_Balance": "AccountBalance","Credit_Score": "CreditScore","Tenure_Years": "Tenure",
    "Join_Date": "CustomerJoinDate","Risk_Level": "RiskLevel","Monthly_Spend": "MonthlySpend","Satisfaction_Score": "SatisfactionScore"})
df = df.drop(columns=["Customer_ID"], errors="ignore")
df["Income_Lakhs"] = df["AnnualIncome"] / 100000
df["MonthlyIncome"] = df["AnnualIncome"] / 12
df["Income_After_Spending"] = (df["MonthlyIncome"] - df["MonthlySpend"])
df["Balance_Income_Ratio"] = (df["AccountBalance"] /df["AnnualIncome"].replace(0, float("nan")))
```
