#### Level 2: Intermediate Iterator Problems

```python
# Problem 6 Create a custom iterator that prints numbers from 1 to 10.

class NumberIterator:
    def __init__(self):
        self.num = 1
    def __iter__(self):
        return self
    def __next__(self):
        if self.num <= 10:
            value = self.num
            self.num += 1
            return value
        else:
            raise StopIteration
itr = NumberIterator()
for i in itr:
    print(i)
```

```python
# Problem 7 Create a custom iterator that prints only even numbers from 2 to 20.

class EvenIterator:
    def __init__(self):
        self.num = 2
    def __iter__(self):
        return self
    def __next__(self):
        if self.num <= 20:
            value = self.num
            self.num += 2
            return value
        else:
            raise StopIteration
itr = EvenIterator()
for i in itr:
    print(i)
```

```python
# Problem 8 Create a custom iterator that prints numbers in reverse order from 10 to 1.

class ReverseIterator:
    def __init__(self):
        self.num = 10
    def __iter__(self):
        return self
    def __next__(self):
        if self.num >= 1:
            value = self.num
            self.num -= 1
            return value
        else:
            raise StopIteration
itr = ReverseIterator()
for i in itr:
    print(i)
```

```python
# Problem 9 Create a custom iterator that prints the characters of the word "GENERATOR" one by one.

class WordIterator:
    def __init__(self):
        self.word = "GENERATOR"
        self.index = 0
    def __iter__(self):
        return self
    def __next__(self):
        if self.index < len(self.word):
            ch = self.word[self.index]
            self.index += 1
            return ch
        else:
            raise StopIteration
itr = WordIterator()
for ch in itr:
    print(ch)
```

```python
#Problem 10 Create a custom iterator that prints the multiplication table of 5.
#Expected Output:
#5
#10
#15
#20
#25
#30
#35
#40
#45
#50

class TableIterator:
    def __init__(self):
        self.num = 1
    def __iter__(self):
        return self
    def __next__(self):
        if self.num <= 10:
            value = 5 * self.num
            self.num += 1
            return value
        else:
            raise StopIteration
itr = TableIterator()
for i in itr:
    print(i)
```

### Level 4: Generator with Lists

```python
# Problem 16 Given the following list:
#numbers = [10, 15, 20, 25, 30, 35]
#Create a generator that yields only the even numbers.

numbers = [10,15,20,25,30,35]
def even_generator(nums):
    for i in nums:
        if i % 2 == 0:
            yield i
gen = even_generator(numbers)
for num in gen:
    print(num)
```

```python
# Problem 17 Create a generator that yields only the odd numbers from the list below.
#numbers = [1, 4, 7, 10, 13, 16, 19]

numbers = [1, 4, 7, 10, 13, 16, 19]
def odd_generator(nums):
    for i in nums:
        if i % 2 != 0:
            yield i
gen = odd_generator(numbers)
for num in gen:
    print(num)
    
```

```python
# Problem 18 Given a list of names, create a generator that yields one name at a time.
#names = ["Alice", "Bob", "Charlie", "David", "Emma"]

names = ["Alice", "Bob", "Charlie", "David", "Emma"]
def name_generator(names):
    for name in names:
        yield name
gen = name_generator(names)
for name in gen:
    print(name)
```

```python
# Problem 19 Given a list of numbers, create a generator that yields only numbers greater than 50.
#numbers = [20, 55, 40, 90, 60, 15, 75]

numbers = [20, 55, 40, 90, 60, 15, 75]
def greater_than_50(nums):
    for num in nums:
        if num > 50:
            yield num
gen = greater_than_50(numbers)
for num in gen:
    print(num)
```

```python
# Problem 20 Create a generator that yields the length of each word in the following list.
# words = ["Python", "Java", "AI", "Generator", "Iterator"]

words = ["Python", "Java", "AI", "Generator", "Iterator"]
def length_generator(words):
    for word in words:
        yield len(word)
gen = length_generator(words)
for length in gen:
    print(length)
```

```python
# Problem 21 Create a custom iterator that prints the first 15 natural numbers.

class NaturalIterator:
    def __init__(self):
        self.num = 1
    def __iter__(self):
        return self
    def __next__(self):
        if self.num <= 15:
            value = self.num
            self.num += 1
            return value
        else:
            raise StopIteration
itr = NaturalIterator()
for i in itr:
    print(i)
```

```python
# Problem 23 Create a custom iterator that prints all alphabet letters from A to Z.

class AlphabetIterartor:
    def __init__(self):
        self.ch = 'A'
    def __iter__(self):
        return self
    def __next__(self):
        if self.ch <= 'Z':
            value = self.ch
            self.ch = chr(ord(self.ch)+1)
            return value
        else:
            raise StopIteration
itr = AlphabetIterartor()
for letter in itr:
    print(letter)
```

```python
# Problem 24 Create a custom iterator that prints all vowels from the string below. text = "Artificial Intelligence"

text = "Artificial Intelligence"
class VowelIterator:
    def __init__(self):
        self.text = text
        self.index = 0
    def __iter__(self):
        return self
    def __next__(self):
        vowels = "AEIOUaeiou"
        while self.index < len(self.text):
            ch = self.text[self.index]
            self.index += 1
            if ch in vowels:
                return ch
        raise StopIteration
itr = VowelIterator()
for ch in itr:
    print(ch)
    
```

```python
# Problem 25 Create a custom iterator that prints the Fibonacci sequence up to 10 terms.

class FibonacciIterator:
    def __init__(self):
        self.a = 0
        self.b = 1
        self.count = 0
    def __iter__(self):
        return self
    def __next__(self):
        if self.count < 10:
            value = self.a
            self.a, self.b = self.b, self.a + self.b
            self.count += 1
            return value
        else:
            raise StopIteration
itr = FibonacciIterator()
for num in itr:
    print(num)
        
```

### Level 6: Advanced Generator Problems

```python
# Problem 26 Create a generator that yields the Fibonacci sequence up to 15 terms.

def fibonacci_generator():
    a = 0
    b = 1
    for i in range(15):
        yield a
        a,b = b,a+b
gen = fibonacci_generator()
for num in gen:
    print(num)
```

```python
# Problem 27 Create a generator that yields all prime numbers between 1 and 100.

def prime_generator():
    for num in range(2,101):
        is_prime = True
        for i in range(2,num):
            if num % i == 0:
                is_prime = False
                break
        if is_prime:
            yield num
gen = prime_generator()
for prime in gen:
    print(prime)
```

```python
# Problem 28 Create a generator that yields all factors of a given number.
#Example Input: num = 24

def factor_generator(number):
    for i in range(1,number+1):
        if number % i == 0:
            yield i
gen = factor_generator(num)
for factor in gen:
    print(factor)
```

```python
# Problem 29 Create a generator that yields the countdown from 20 to 1.

def countdown_generator():
    for i in range(20,0,-1):
        yield i
gen = countdown_generator()
for num in gen:
    print(num)
```

```python
# Problem 30 Create an infinite generator that yields multiples of 5. Print only the first 15 values using next().

def multiple_of_five():
    num = 5
    while True:
        yield num
        num += 5
gen = multiple_of_five()
for i in range(15):
    print(next(gen))
```
