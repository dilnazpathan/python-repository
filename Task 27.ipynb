```python
# 17. Store a list of numbers in a file, then read and calculate their sum.

file = open("numbers.txt","w")
numbers = input("enter numbers separated by space:")
file.write(numbers)
file.close()
file = open("numbers.txt","r")
content = file.read()
num_list = content.split()
total = 0
for num in num_list:
    total += int(num)
file.close()
print("sum of numbers",total)
```

```python
# 20. Read a CSV-like data from a text file and print it in tabular form.

file = open("data.txt","r")
for line in file:
    data = line.strip().split(",")
    print(data[0],"\t",data[1],"\t",data[2])
file.close()
```

```python
# 23. Take multiple lines of text from user until they type “STOP”, then save all lines in a file.

file = open("text.txt","w")
print("enter text (type STOP to finish):")
while True:
    line = input()
    if line == "STOP":
        break
    file.write(line + "\n")
file.close()
print("data saved successfully")
    
```

```python
# 24. Write a program to encrypt the content of a file (simple character shifting) and save it in  another file.

file1 = open("student.txt","r")
content = file1.read()
encrypted =""
for ch in content:
    encrypt += chr(ord(ch) + 1)
file2 = open("encrypted.txt","w")
file2.write(encrypted)
file1.close()
file2.close()
print("file encrypted successfully")
```

```python
# 25. Read a file containing student marks, calculate average, and write the result in another file.

file1 = open("marks.txt","r")
total = 0
count = 0
for line in file1:
    total += int(line.strip())
    count += 1
average = total / count
file1.close()
file2 = open("average.txt","w")
file2.write("average marks = "+str(average))
file2.close()
print("average calculated and saved successfully")
```

```python
# 26. Create a program that keeps a record of daily expenses in a file and can show total expenses.

file = open("expenses.txt","a")
expense = input("enter todays expense:")
file.write(expense + "\n")
file.close()
file = open("expenses.txt","r")
total = 0
for line in file:
    total += float(line.strip())
file.close()
print("total expenses =",total)    
```

```python
# 27. Write a program to remove all blank lines from a file.

file = open("student.txt","r")
lines = file.readlines()
file.close()
file = open("student.txt","w")
for  line in lines:
    if line.strip() != "":
        file.write(line)
file.close()
print("blank line removed successfully")
```

```python
# 28. Read a file and write only the unique lines into another file (remove duplicates).

file1 = open("student.txt","r")
lines = file1.readlines()
unique_lines = []
for line in lines:
    if line not in unique_lines:
        unique_lines.append(line)
file1.close()
file2 = open("unique.txt","w")
for line in unique_lines:
    file2.write(line)
file2.close()
print("unique lines written successfully")
```

```python
# 29. Create a menu-driven program:
#• Write to file
#• Read file
#• Append to file
#• Exit

while True:
    print("\n------MENU------")
    print("1. write to file")
    print("2. read file")
    print("3. append to file")
    print("4. exit")
    choice = int(input("enter your choice:"))
    if choice == 1:
        file = open("student.txt","w")
        data = input("enter text: ")
        file.write(data)
        file.close()
        print("data written successfully")
    elif choice == 2:
        file = open("student.txt","r")
        print("\nFile content:")
        print(file.read())
        file.close()
    elif choice == 3:
        file = open("student.txt","a")
        data = input("enter text to append:")
        file.write("\n" + data)
        file.close()
        print("data appended successfully")
    elif choice == 4:
        print("program exited")
        break
    else:
        print("invalid choice! please try again")
```
