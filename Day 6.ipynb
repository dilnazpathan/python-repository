```python
# question 25 Given data = (10, 20, 30, 20, 40, 20, 50), convert it into a dictionary where key is the number and value is its count of occurrences
#(using tuple methods + dictionary)

data = (10,20,30,40,20,50)
result = {}
for i in data:
    result[i] = data.count(i)
print(result)
```
