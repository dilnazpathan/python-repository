### Python Lambda Function

```python
# question 1 Write a lambda function to return the square of a number.
#Example:
#Input: 5
#Output: 25

result= lambda x :x*x
result(5)
```

```python
# question 5 Write a lambda function that returns "Even" if a number is even; otherwise "Odd".
#Input: 9
#Output: Odd

even = lambda x : "even" if x % 2 == 0 else "odd"
print(even(9))
```

```python
# question 6 Use map() with lambda to square every element of a list.
#Input:[1,2,3,4]
#Output:[1,4,9,16]

list1 =[1,2,3,4]
square = list(map(lambda x: x * x,list1))
print(square)
                        
```

```python
# question 7 Use filter() with lambda to keep only even numbers.
#Input:[1,2,3,4,5,6]
#Output:[2,4,6]

numbers = [1,2,3,4,5,6]
result = list(filter(lambda x: x % 2 == 0, numbers))
print(result)
```

```python
# question 29 Sort a list of dictionaries by age ascending and then salary descending.
#employees = [
#{"name":"Ram","age":25,"salary":50000},
#{"name":"Amit","age":25,"salary":60000},
#{"name":"Neha","age":23,"salary":45000},
#{"name":"Raj","age":27,"salary":55000}
#]

employees = [
{"name":"Ram","age":25,"salary":50000},
{"name":"Amit","age":25,"salary":60000},
{"name":"Neha","age":23,"salary":45000},
{"name":"Raj","age":27,"salary":55000}
]
result = sorted(employees,key= lambda x:(x["age"],-x["salary"]))
print(result)
```

```python
# question 30 Given:
#students = [
#{"name":"Ram","marks":[70,80,90]},
#{"name":"Amit","marks":[90,92,94]},
#{"name":"Neha","marks":[60,65,70]},
#{"name":"Raj","marks":[88,84,90]}
#]
#Using only lambda with built-in functions (map, filter, sorted, max, min, reduce where appropriate):
#1. Calculate each student's average marks. 

students = [
{"name":"Ram","marks":[70,80,90]},
{"name":"Amit","marks":[90,92,94]},
{"name":"Neha","marks":[60,65,70]},
{"name":"Raj","marks":[88,84,90]}
]
average = list(map(lambda s:(s["name"], sum(s["marks"])/len(s["marks"])),students))
print("Average:",average)
```

```python
# 2. Sort students by average marks (highest first).

students = [
{"name":"Ram","marks":[70,80,90]},
{"name":"Amit","marks":[90,92,94]},
{"name":"Neha","marks":[60,65,70]},
{"name":"Raj","marks":[88,84,90]}
]
result = sorted(students,key = lambda x:sum(x["marks"])/len(x["marks"]),reverse = True)
print(result)
```

```python
# 3. Find the topper.

topper = max(students,key = lambda s : sum (s["marks"])/len(s["marks"]))
print("topper:",topper)
```

```python
# 4. Filter students whose average is greater than 80.

filtered = list(filter(lambda s:sum(s["marks"])/len(s["marks"])>80,students))
print(filtered)                
                
```

```python
#5. Print only the names of the filtered students.

result = list(map(lambda s: s["name"],filtered))
print(result)
```
