```python
# Q30 Real Industry Mini Project
#Your manager asks you to prepare a cleaned dataset for the Machine Learning team.
#Perform all of these steps:
#1. Load the dataset.
#2. Display dataset information.
#3. Check missing values.
#4. Remove duplicate rows.
#5. Rename the following columns:
#o CustomerId → Customer_ID
#o CreditScore → Credit_Score
#o EstimatedSalary → Salary
#o Exited → Target
#6. Convert:
#o Age → float
#o Balance → float
# Target → string
#7. Replace missing values:
#o Age → median
#o Balance → mean
#8. Remove:
#o RowNumber
#o Surname
#9. Display:
#o First 10 rows
#o Statistical summary
#o Value counts of Target
#o Final dataset information

import pandas as pd


df = pd.read_excel(r"C:\Users\DELL\Downloads\Bank_customer_churn.xlsx")

print("========== Dataset Information ==========")
df.info()


print("\n========== Missing Values ==========")
print(df.isnull().sum())


df.drop_duplicates(inplace=True)

df.rename(columns={
    "CustomerId": "Customer_ID",
    "CreditScore": "Credit_Score",
    "EstimatedSalary": "Salary",
    "Exited": "Target"
}, inplace=True)


df["Age"] = df["Age"].astype(float)
df["Balance"] = df["Balance"].astype(float)
df["Target"] = df["Target"].astype(str)


df["Age"] = df["Age"].fillna(df["Age"].median())
df["Balance"] = df["Balance"].fillna(df["Balance"].mean())

df.drop(columns=["RowNumber", "Surname"], inplace=True)


print("\n========== First 10 Rows ==========")
print(df.head(10))

print("\n========== Statistical Summary ==========")
print(df.describe())

print("\n========== Target Value Counts ==========")
print(df["Target"].value_counts())

print("\n========== Final Dataset Information ==========")
df.info()
```
