### Seaborn catplot() — 30 Questions

```python
# A. Titanic Dataset — 15 Questions
#Use:
#import seaborn as sns
#titanic = sns.load_dataset("titanic")


import seaborn as sns
titanic = sns.load_dataset("titanic")
titanic
```

```python
#1. Survival by Gender
#Use catplot() to show the number of passengers who survived and did not survive for each
#gender.
#• X-axis → sex
#• Hue → survived
#• Chart → count

import matplotlib.pyplot as plt
sns.catplot(
    data = titanic,
    x = "sex",
    hue = "survived",
    kind = "count"
)
plt.title("survival by gender")
plt.xlabel("gender")
plt.ylabel("number of passeneger")
plt.show()
```

```python
# 2. Survival by Passenger Class
#Create a categorical plot showing the number of passengers in each class.
#Use:
#• x="class"
#• kind="count"

sns.catplot(
    data =titanic,
    x = "class",
    kind = "count"
)
plt.title("number of passeneger by class")
plt.xlabel("passenger class")
plt.ylabel("number of passenger")
plt.show()
```

```python
# 3. Survival Count by Class
#Show the number of survivors and non-survivors for each passenger class.
#Use:
#x="class"
#hue="survived"
#kind="count"

sns.catplot(
    data = titanic,
    x = "class",
    hue = "survived",
    kind = "count"
)
plt.title("survived count by passenger class")
plt.xlabel("passenegr class")
plt.ylabel("number of passenger")
plt.show()
```

```python
# 4. Survival Percentage by Gender
#Create a categorical visualization comparing survival across sex.
#Use catplot() and choose an appropriate categorical plot.
#Then identify which gender has a higher survival rate.

survival_rate = titanic.groupby("sex")["survived"].mean().reset_index()
survival_rate["survived"] = survival_rate["survived"]*100
sns.catplot(
    data = survival_rate,
    x="sex",
    y="survived",
    kind = "bar"
)
plt.title("survived percentage by gender")
plt.xlabel("gender")
plt.ylabel("survival percentage (%) ")
plt.show()
```

```python
# 5. Age Distribution by Gender
#Use catplot() to compare passenger age across different genders.
#Try an appropriate kind for showing the distribution.

sns.catplot(
    data = survival_rate,
    x="sex",
    y="survived",
    kind = "box"
)
plt.title("age distribution by gender")
plt.xlabel("gender")
plt.ylabel("age")
plt.show()
```

```python
# 6. Age Distribution by Passenger Class
#Create a categorical plot showing age distribution for:
#• First class
#• Second class
#• Third class

sns.catplot(
    data = titanic,
    x="class",
    y="age",
    kind = "box"
)
plt.title("age distribution by passenger class")
plt.xlabel("passenger class")
plt.ylabel("age")
plt.show()
```

```python
# 7. Fare by Passenger Class
#Use catplot() to compare fare across the three passenger classes.
#Add:
#• Appropriate color
#• Title
#• X/Y labels

sns.catplot(
    data = titanic,
    x="class",
    y="fare",
    kind="box",
    color="skyblue"
)
plt.title("fare by passenger class")
plt.xlabel("passenger class")
plt.ylabel("fare")
plt.show()
```

```python
# 8. Fare Distribution by Gender
#Create a categorical visualization of fare for male and female passengers.
#Identify whether the fare distributions appear different.


sns.catplot(
    data = titanic,
    x="sex",
    y="fare",
    kind="box",
    color="lightgreen"
)
plt.title("fare distribution by gender")
plt.xlabel("gender")
plt.ylabel("fare")
plt.show()
```

```python
# 9. Age vs Passenger Class
#Use a categorical plot to compare the distribution of age across class.
#Add sex using hue.

sns.catplot(
    data=titanic,
    x="class",
    y="age",
    hue="sex",
    kind="box"
)

plt.title("Age Distribution by Passenger Class and Gender")
plt.xlabel("Passenger Class")
plt.ylabel("Age")
plt.show()
```

```python
# 10. Survival by Class and Gender
#Create a categorical visualization showing survival for each passenger class and gender.
#Use:
#x="class"
#hue="sex"
#Use an appropriate kind.

sns.catplot(
    data = titanic,
    x="class",
    hue="sex",
    kind = "count"
)
plt.title("passenger count by class and gender")
plt.xlabel("passenger class")
plt.ylabel("number of passenger")
plt.show()
```

```python
# 11. Embark Town vs Survival
#Create a count plot using catplot() showing passenger counts for each embark_town.
#Use survived as hue.

sns.catplot(
    data = titanic,
    x="embark_town",
    hue="survived",
    kind="count"
)
plt.title("embrak town vs survival")
plt.xlabel("embark town")
plt.ylabel("number of passenger")
plt.show()
```

```python
# 12. Passenger Class vs Fare with Gender
#Create a categorical visualization comparing fare across passenger classes.
#Use:
#x="class"
#y="fare"
#hue="sex"
#Choose an appropriate kind.

sns.catplot(
    data = titanic,
    x="class",
    y="fare",
    hue="sex",
    kind="box"
)
plt.title("fare by passenger")
plt.xlabel("passenger class")
plt.ylabel("fare")
plt.show()
```

```python
# 13. Survival vs Age Group
#Create an age_group column:
#• Child → Age < 18
#• Adult → 18–59
#• Senior → 60+
#Use catplot() to compare survival across these groups.

import pandas as pd
titanic["age_group"] = pd.cut(
    titanic["age"],
    bins=[0,18,60,float("inf")],
    labels=["child","adult","senior"],
    right=False
)
sns.catplot(
    data = titanic,
    x="age_group",
    hue="survived",
    kind="count"
)
plt.title("survival by age group")
plt.xlabel("age group")
plt.ylabel("number of passeger")
plt.show()
```

```python
# 14. Class, Gender and Survival
#Create a categorical visualization where:
#• X → class
#• Y → survival-related measure
#• Hue → sex
#• Columns → survived
#Use col= to create separate panels.

sns.catplot(
    data=titanic,
    x="class",
    hue="sex",
    col="survived",
    kind="count"
)
plt.show()
```

```python
# 15. Titanic Multi-Panel Analysis
#Create a catplot() with:
#• X → class
#• Y → fare
#• Hue → sex
#• Columns → survived
#Use an appropriate categorical kind.
#Add:
#• Figure size
#• Title
#• Color palette
#• Appropriate labels
#Explain what you observe.

sns.catplot(
    data=titanic,
    x="class",
    y="fare",
    hue="sex",
    col="survived",
    kind="box",
    height=5,
    aspect=1.1,
    palette="Set2"
)
plt.subplots_adjust(top=0.85)
plt.suptitle("titanic fare analysis by class, gender and survival")
plt.show()
```
