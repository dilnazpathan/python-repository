```python
# question 32 (Hard) Create class BankAccount with private __balance. Implement setter for balance with these rules:
#• Balance cannot go below minimum balance (1000).
#• If balance > 100000, give 5% bonus automatically.
#• Log every update (use a class list to store history). Add getter and class method to show total accounts created.

class BankAccount:
    total_accounts = 0     
    history = []            
    def __init__(self, balance):
        self.__balance = balance
        BankAccount.total_accounts += 1
    def get_balance(self):
        return self.__balance
    def set_balance(self, amount):
        if amount < 1000:
            print("Minimum balance is 1000")
        else:
            if amount > 100000:
                amount = amount + (amount * 0.05)
                print("5% Bonus Added")
            self.__balance = amount
            BankAccount.history.append(amount)
            print("Balance Updated")

    @classmethod
    def show_total_accounts(cls):
        return cls.total_accounts
b1 = BankAccount(5000)
b1.set_balance(120000)
print("Balance:", b1.get_balance())
print("History:", BankAccount.history)
print("Accounts:", BankAccount.show_total_accounts())
```
