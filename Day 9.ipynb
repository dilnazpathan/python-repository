```python
# question 17 Write a program to create a simple ATM system:
# o Check balance
# o Withdraw money
# o Deposit money
# o Invalid option message

balance = 10000
print("1. check balance")
print("2. Withdraw")
print("3. Deposite")
choice = int(input("enter your choice"))
if choice == 1:
    print("balance:",balance)
elif choice == 2:
    amount = float (input("enter amount to withdraw:"))
    if amount <= balance:
        balance -= amount
        print("Remaining Balance:", balance)
    else:
        print("Insufficent Balance")
elif choice == 3:
    amount = float(input("enter amount to deposite:"))
    balance <= amount
    print("updated balance =",balance)
else:
    print("Invalid option")
```

```python
# question 18 Write a program to calculate income tax:
# o Income below ₹250000 → No tax
# o ₹250000–₹500000 → 5%
# o ₹500000–₹1000000 → 20%
# o Above ₹1000000 → 30%

income = float(input("enter annual income:"))
if income < 250000:
    tax = 0
elif income <= 500000:
    tax = income * 0.05
elif income <= 1000000:
    tax = income * 0.20
else:
    tax = income * 0.30
print("Tax = ",tax)
```

```python
# question 19 Write a program to determine the type of roots in a quadratic equation using discriminant.

a = float(input("enter a: "))
b = float(input("enter b: "))
c = float(input("enter c: "))
d = b**2 - 4*a*c
if d>0:
    print("two real and distinct roots")
elif d == 0:
    print("Two real and equal roots")
else:
    print("Imaginary roots")
```
