```python
# question 1 Create two parent classes Father and Mother. Create a child class Child that inherits from both.

class Father:
    def father_info(self):
        print("father name: rahul")
        print("father profession:engineer")
class Mother:
    def mother_info(self):
        print("mother name: anjali")
        print("mother profession: teacher")
class Child(Father,Mother):
    def child_info(self):
        print("child name: priya")
c = Child()
c.father_info()
c.mother_info()
c.child_info()
        
```

```python
# question 2 Define class A with method show() and class B with method display(). Create class C that inherits from both and calls both methods.

class A:
    def show(self):
        print("this method show() from class A")
class B:
    def display(self):
        print("this method display() from class B")
class C(A,B):
    def call_method(self):
        self.show()
        self.display()
obj = C()
obj.call_method()
```

```python
# question 3 Create classes Teacher and Author. Create class Person that inherits from both and has attributes from both.

class Teacher:
    def __init__(self,subject):
        self.subject = subject
class Author:
    def __init__(self,book):
        self.book = book
class Person(Teacher,Author):
    def __init__(self,name,subject,book):
        self.name = name
        Teacher.__init__(self,subject)
        Author.__init__(self,book)
    def display_info(self):
        print("name:",self.name)
        print("subject:",self.subject)
        print("book:",self.book)
p = Person("ravi","python programming","learning programming")
p.display_info()
```

```python
# question 11 Create two parent classes with static methods. Create a child class that inherits from both and calls the static methods.

class Father:
    @staticmethod
    def father_info():
        print("father stactic method")
class Mother:
    @staticmethod
    def mother_info():
        print("mother static method")
class Child(Father,Mother):
    
    def child_info(self):
        print("child class method")
c = Child()
c.father_info()
c.mother_info()
c.child_info()
```

```python
# question 12 Define class MathUtils with static method add(). Create class AdvancedMath that inherits and uses the static method.

class MathUtils:
    @staticmethod
    def add(a,b):
        return a + b
class AdvanceMath(MathUtils):
    def calculate(self):
        result = self.add(10,20)
        print("addition result:",result)
a = AdvanceMath()
a.calculate()
```

```python
# question 13 Create class Validator with static method is_valid(). Create class Form that inherits and uses it.

class Validator:
    @staticmethod
    def is_valid(value):
        return bool(value)
class Form(Validator):
    def submit(self,value):
        if self.is_valid(value):
            return "form submitted successfully"
        return "valid input"
form = Form()
print(form.submit("john"))
print(form.submit(" "))
```

```python
# question 14 Define class Converter with static method to_celsius(). Create class WeatherApp that inherits from multiple classes including Converter.

class Converter:
    @staticmethod
    def to_celsius(fahrenheit):
        return (fahrenheit - 32) * 5/9
class Location:
    def show_location(self):
        return "location: New York"
class WheatherApp(Converter,Location):
    def display_weather(self,fahrenheit):
        celsius = self.to_celsius(fahrenheit)
        print(self.show_location())
        print(f"temperature: {celsius:2f}C")
app = WheatherApp()
app.display_weather(86)
```

```python
# question 15 Create class Logger with static method log(). Create class Application that inherits from Logger and another class.

class Logger:
    @staticmethod
    def log(message):
        print("log: message")
class User:
    def login(self,username):
        print(f"{username} logged in")
class Application(Logger,User):
    def start(self,username):
        self.log("application started")
        self.login(username)
app = Application()
app.start("john")
```

```python
# question 17 Create class Calculator with static method and another class Display. Create child class that uses both.

class Calculator:
    @staticmethod
    def add(a,b):
        return a+b
class Display:
    def show(self,result):
        print("result:",result)
class MyApp(Calculator,Display):
    def calculate(self,a,b):
        result = self.add(a,b)
        self.show(result)
app = MyApp()
app.calculate(10,20)
```

### Multiple Inheritance with Abstract Methods (19-25)

```python
# question 19 Create two abstract classes Shape and Color. Create a concrete class that inherits from both and implements abstract methods.
from abc import ABC,abstractmethod
class Shape(ABC):
    @abstractmethod
    def draw(self):
        pass
class Color(ABC):
    @abstractmethod
    def fill(self):
        pass
class Circle(Shape,Color):
    def draw(self):
        print("drawing a circle")
    def fill(self):
        print("filling the circle with red color")
c = Circle()
c.draw()
c.fill()
```

```python
# question 20 Define abstract class Payment and abstract class Notification. Create class OnlinePayment that inherits from both.

from abc import ABC,abstractmethod
class Payment(ABC):
    @abstractmethod
    def make_payment(self,amount):
        pass
class Notification(ABC):
    @abstractmethod
    def send_notification(self):
        pass
class OnlinePayment(Payment,Notification):
    def make_payment(self,amount):
        print(f"payment of ${amount} successful")
    def send_notification(self):
        print("payment notification sent to the user")
obj = OnlinePayment()
obj.make_payment(500)
obj.send_notification()
```

```python
# question 21 Create abstract class Vehicle and abstract class Electric. Create class ElectricCar that inherits from both.

from abc import ABC,abstractmethod
class Vehicle:
    @abstractmethod
    def start(self):
        pass
class Electric(ABC):
    @abstractmethod
    def charge(self):
        pass
class ElectricCar(Vehicle,Electric):
    def start(self):
        print("electric car started")
    def charge(self):
        print("electric car is charging")
car = ElectricCar()
car.start()
car.charge()
```

```python
# question 22 Define abstract class Readable and abstract class Writable. Create class File that implements both.

from abc import ABC,abstractmethod
class Readable(ABC):
    @abstractmethod
    def read(self):
        pass
class Writeable(ABC):
    @abstractmethod
    def write(self,data):
        pass
class File(Readable,Writeable):
    def read(self):
        print("reading a file")
    def write(self,data):
        print(f"writing data from file: {data}")
f = File()
f.read()
f.write("hello world")
```

```python
# question 26 (Problem) Create two parent classes:
#• Employee with method get_salary()
#• Person with method get_details() Create class Manager that inherits from both. Add a method that combines both parent methods.

class Employee:
    def get_salary(self):
        return 50000
class Person:
    def get_details(self):
        return "Name: John, Age: 30"
class Manager(Employee, Person):
    def show_manager_info(self):
        details = self.get_details()
        salary = self.get_salary()
        print(details)
        print("Salary:", salary)
manager = Manager()
manager.show_manager_info()
```

```python
# question 28 (Problem) Design a system using multiple inheritance:
#• Class Writer
#• Class Editor Create class Author that inherits from both. Add static method to count total authors.

class Writer:
    def __init__(self,name):
        self.name = name
    def write(self):
        print(self.name, "is writing")
class Editor:
    def edit(self):
        print(self.name, "is editing")
class Author(Writer,Editor):
    total_authors = 0
    def __init__(self,name):
        super().__init__(name)
        Author.total_authors += 1
    @staticmethod
    def count_authors():
        print("total authors:", Author.total_authors)
author1 = Author("dilnaz")
author2 = Author("tanmayi")
author3 = Author("ruchika")
author1.write()
author2.edit()
Author.count_authors()
```

```python
# question 29 (Problem) Create two abstract classes Printable and Scannable. Create class
#PrinterScanner that inherits from both and implements both abstract methods.

from abc import ABC, abstractmethod
class Printable(ABC):
    @abstractmethod
    def print_document(self):
        pass
class Scannable(ABC):
    @abstractmethod
    def scan_document(self):
        pass
class PrinterScanner(Printable, Scannable):
    def print_document(self):
        print("Printing document...")
    def scan_document(self):
        print("Scanning document...")
device = PrinterScanner()
device.print_document()
device.scan_document()
```
