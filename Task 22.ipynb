### Basic Hierarchical Inheritance (1–12)

```python
# 1. Person → Student, Teacher
#Question
#Create a parent class Person with attributes name and age. Create two child classes Student and Teacher that inherit from Person.
#Concept
#• Hierarchical Inheritance
#• Parent attributes inherited by multiple child classes.
#How to Solve
#1. Create a parent class Person.
#2. Add name and age using a constructor.
#3. Create Student(Person).
#4. Create Teacher(Person).
#5. Create objects of both classes.
#6. Show that both children can access name and age.

class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age
class Student(Person):
    def display_student(self):
        print("student name:", self.name)
        print("student age:",self.age)
class Teacher(Person):
    def display_teacher(self):
        print("teacher name:",self.name)
        print("teacher age:", self.age )
std1 = Student("alice",22)
teacher = Teacher("priya",29)
std1.display_student()
teacher.display_teacher()
    
```

```python
# 2. Animal → Dog, Cat
#Question
#Define class Animal with method eat(). Create subclasses Dog and Cat.
#Concept
#• Method inheritance.
#How to Solve
#1. Create parent class Animal.
#2. Write method eat().
#3. Create child classes Dog and Cat.
#4. Create objects.
#5. Call eat() using both objects.

class Animal:
    def eat(self):
        print("animal is eating")
class Dog(Animal):
    pass
class Cat(Animal):
    pass
dog = Dog()
cat = Cat()
print("dog:",)
dog.eat()
print("\n cat:")
cat.eat()
```

```python
# 3. Vehicle → Car, Bike
#Question
#Create class Vehicle with method start(). Create subclasses Car and Bike.
#Concept
#• One parent method shared by multiple children.
#How to Solve
#1. Create Vehicle.
#2. Add start().
#3. Inherit it in Car.
#4. Inherit it in Bike.
#5. Call start() from both child objects.

class Vehicle:
    def start(self):
        print("vehicle is starting")
class Car(Vehicle):
    pass
class Bike(Vehicle):
    pass
c = Car()
print("car:",)
c.start()
b = Bike()
print("bike:",)
b.start()
```

```python
# 4. Shape → Circle, Rectangle
#Question
#Define class Shape with method area(). Create subclasses Circle and Rectangle.
#Concept
#• Parent provides common method.
#• Children may later override it.
#How to Solve
#1. Create Shape.
#2. Add area().
#3. Create Circle.
#4. Create Rectangle.
#5. Call area().


class Shape:
    def area(self):
        print("Calculating area of a shape.")
class Circle(Shape):
    pass
class Rectangle(Shape):
    pass
circle1 = Circle()
rectangle1 = Rectangle()
print("Circle:")
circle1.area()
print("\nRectangle:")
rectangle1.area()
```

```python
# 13. Person Constructor
#Concept
#Using super() to call the parent constructor.
#How to Solve
#1. Create constructor in parent.
#2. Create child constructors.
#3. Use super().__init__().
#4. Add child-specific attribute.
#5. Display all data.

class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age
class Student(Person):
    def __init__(self,name,age,roll_no):
        super().__init__(name,age)
        self.roll_no = roll_no
    def display(self):
        print("name:", self.name)
        print("age:",self.age)
        print("roll no:",self.roll_no)
s = Student("dilnaz",23,101)
s.display()
```

```python
# 14. Animal Sound Override
#Concept
#Method Overriding.
#How to Solve
#1. Parent has sound().
#2. Dog overrides.
#3. Cat overrides.
#4. Call method from each object.


class Animal:
    def sound(self):
        print("Animal makes a sound.")
class Dog(Animal):
    def sound(self):
        print("Dog barks.")
class Cat(Animal):
    def sound(self):
        print("Cat meows.")
animal = Animal()
dog = Dog()
cat = Cat()
print("Animal:")
animal.sound()
print("\nDog:")
dog.sound()
print("\nCat:")
cat.sound()
```

```python
# 15. Vehicle Constructor
#Concept
#Constructor inheritance.
#How to Solve
#1. Parent stores brand.
#2. Car adds doors.
#3. Bike adds engine_cc.
#4. Use super().

class Vehicle:
    def __init__(self,brand):
        self.brand = brand
class Car(Vehicle):
    def __init__(self,brand,doors):
        super().__init__(brand)
        self.doors = doors
    def display_car(self):
        print("car brand:",self.brand)
        print("car doors:",self.doors)
class Bike(Vehicle):
    def __init__(self,brand,engine_cc):
        super().__init__(brand)
        self.engine_cc = engine_cc
    def display_bike(self):
        print("bike brand:",self.brand)
        print("Engine cc:",self.engine_cc)
c = Car("Mercedes",4)
c.display_car()
b = Bike("Apache",150)
b.display_bike()
```

```python
# 16. Shape Area
#Concept
#Polymorphism using overriding.
#How to Solve
#1. Parent constructor.
#2. Circle calculates πr2.
#3. Square calculates side2.
#4. Override area().


class Shape:
    def __init__(self, name):
        self.name = name
    def area(self):
        print("Area of shape.")
class Circle(Shape):
    def __init__(self, name, radius):
        super().__init__(name)
        self.radius = radius
    def area(self):
        print("Area of Circle:", 3.14 * self.radius * self.radius)
class Square(Shape):
    def __init__(self, name, side):
        super().__init__(name)
        self.side = side
    def area(self):
        print("Area of Square:", self.side * self.side)
circle = Circle("Circle", 5)
square = Square("Square", 4)
circle.area()
square.area()
```

```python
# 18. BankAccount Interest
#Concept
#Different child behavior.
#How to Solve
#1. Parent constructor.
#2. Savings has one interest.
#3. Current has another.
#4. Calculate separately.


class BankAccount:
    def __init__(self, balance):
        self.balance = balance
    def calculate_interest(self):
        print("Calculating interest.")
class SavingsAccount(BankAccount):
    def calculate_interest(self):
        interest = self.balance * 0.05
        print("Savings Account Interest:", interest)
class CurrentAccount(BankAccount):
    def calculate_interest(self):
        interest = self.balance * 0.03
        print("Current Account Interest:", interest)
savings = SavingsAccount(10000)
current = CurrentAccount(10000)
savings.calculate_interest()
current.calculate_interest()
```

```python
# 24. Bank System
#Concept
#Three children inherit one parent.
#How to Solve
#1. Parent → Account
#2. Store balance.
#3. Savings → interest.
#4. Current → overdraft.
#5. FixedDeposit → maturity.
#6. Show each separately.


class Account:
    def __init__(self, balance):
        self.balance = balance
class Savings(Account):
    def __init__(self, balance, interest):
        super().__init__(balance)
        self.interest = interest
    def show_savings(self):
        print("Savings Account")
        print("Balance:", self.balance)
        print("Interest:", self.interest)
class Current(Account):
    def __init__(self, balance, overdraft):
        super().__init__(balance)
        self.overdraft = overdraft
    def show_current(self):
        print("Current Account")
        print("Balance:", self.balance)
        print("Overdraft Limit:", self.overdraft)
class FixedDeposit(Account):
    def __init__(self, balance, maturity):
        super().__init__(balance)
        self.maturity = maturity
    def show_fixeddeposit(self):
        print("Fixed Deposit Account")
        print("Balance:", self.balance)
        print("Maturity Amount:", self.maturity)
savings = Savings(50000, 5)
current = Current(30000, 10000)
fixed = FixedDeposit(100000, 120000)
savings.show_savings()
print()
current.show_current()
print()
fixed.show_fixeddeposit()
```

```python
# 27. Employee Salary
#Concept
#Different salary calculations.
#How to Solve
#1. Parent stores common data.
#2. FullTime calculates monthly salary.
#3. PartTime calculates hourly salary.
#4. Override calculation method.

class Employee:
    def __init__(self,name):
        self.name = name
    def calculate_salary(self):
        print("calculating employee salary")
class FullTime(Employee):
    def __init__(self,name,monthly_salary):
        super().__init__(name)
        self.monthly_salary = monthly_salary
    def calculate_salary(self):
        print("full time employee:",self.name)
        print("monthly salary:",self.monthly_salary)
class PartTime(Employee):
    def __init__(self,name,hours,hourly_rate):
        super().__init__(name)
        self.hours = hours
        self.hourly_rate = hourly_rate
    def calculate_salary(self):
        salary = self.hours * self.hourly_rate
        print("part time employee:",self.name)
        print("hourly salary:",salary)
fulltime = FullTime("dilnaz",50000)
parttime = PartTime("ruchika",40,20)
fulltime.calculate_salary()
print()
parttime.calculate_salary()
    
```

```python
# 29. Notification System
#Concept
#Method overriding.
#How to Solve
#1. Parent → Notification
#2. Method send()
#3. Email overrides.
#4. SMS overrides.
#5. Push overrides.
#6. Call send() using each object.


class Notification:
    def send(self):
        print("Sending notification.")
class Email(Notification):
    def send(self):
        print("Sending notification through Email.")
class SMS(Notification):
    def send(self):
        print("Sending notification through SMS.")
class Push(Notification):
    def send(self):
        print("Sending notification through Push Notification.")
email = Email()
sms = SMS()
push = Push()
print("Email:")
email.send()
print("\nSMS:")
sms.send()
print("\nPush:")
push.send()
```
