```python
# 11.Create a monthly sales chart and identify the maximum-sales month. Highlight that point using annotate().

import matplotlib.pyplot as plt
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

sales = [25000, 30000, 28000, 35000, 40000, 38000,
         45000, 42000, 50000, 48000, 55000, 60000]
max_sales = max(sales)
max_index = sales.index(max_sales)
max_month = months[max_index]
plt.plot(months,sales,marker="o")
plt.scatter(max_month,max_sales)
plt.annotate(
    f"highest sales: {max_sales}",
    xy = (max_month, max_sales),
    xytext = (-80,30),
    textcoords="offset points",
     arrowprops=dict(arrowstyle="->")
 )
plt.title("monthly sales")
plt.xlabel("month")
plt.ylabel("sales")
plt.grid(True)
plt.show()

```

```python
# 12.Create a scatter plot of Age vs Income. Highlight customers whose income is greater than ₹1,00,000 using a different marker.

import matplotlib.pyplot as plt

age = [22, 25, 28, 30, 32, 35, 38, 40, 42, 45]
income = [50000, 75000, 90000, 110000, 80000, 120000, 95000, 150000, 130000, 70000]

normal_age = []
normal_income = []
high_age = []
high_income=[]
for a, i in zip(age,income):
    if i > 1000000:
        high_age.append(a)
        high_income.append(i)
    else:
        normal_age.append(a)
        normal_income.append(i)
plt.scatter(normal_age,normal_income,marker="o",label="income<=₹100000")
plt.scatter(high_age, high_income, marker="*", label="Income > ₹1,00,000")
plt.title("Age vs Income")
plt.xlabel("Age")
plt.ylabel("Income (₹)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# 13.Create a stacked bar chart showing monthly sales from:
#o Online
#o Offline
#o Corporate

import matplotlib.pyplot as plt
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
online = [30000, 35000, 40000, 45000, 50000, 55000]
offline = [20000, 25000, 22000, 28000, 30000, 32000]
corporate = [15000, 18000, 20000, 22000, 25000, 28000]
plt.bar(months, online, label="Online")
plt.bar(months, offline, bottom=online, label="Offline")
plt.bar(
    months,
    corporate,
    bottom=[online[i]+offline[i] for i in range (len(months))],
    label="corporate"
)
plt.title("Monthly Sales by Channel")
plt.xlabel("Month")
plt.ylabel("Sales (₹)")
plt.legend()
plt.grid(axis="y")

plt.show()
```

```python
# 14.Create a graph containing actual sales and target sales for 12 months. Highlight the months where actual sales are below the target.

import matplotlib.pyplot as plt
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
actual_sales = [25000, 30000, 28000, 35000, 32000, 40000,
                45000, 38000, 50000, 48000, 55000, 60000]
target_sales = [30000, 30000, 32000, 33000, 35000, 38000,
                42000, 42000, 48000, 50000, 52000, 58000]
plt.plot(months, actual_sales, marker="o", label="Actual Sales")
plt.plot(months, target_sales, marker="s", linestyle="--", label="Target Sales")
for i in range(len(months)):
    if actual_sales[i] < target_sales[i]:
        plt.scatter(months[i],actual_sales[i],marker="x",s=100)
plt.title("Actual Sales vs Target Sales")
plt.xlabel("Month")
plt.ylabel("Sales (₹)")
plt.legend()
plt.grid(True)

plt.show()
```

```python
# 15.Create a mini analytics dashboard using Matplotlib with 4 subplots:
#o Monthly revenue line chart
#o Category-wise sales bar chart
#o Customer-age histogram
#o Revenue vs discount scatter plot
#Add titles, labels, legends, grids, and appropriate figure sizing to make the dashboard readable.

import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Data
# -----------------------------

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

revenue = [50000, 55000, 48000, 62000, 68000, 72000,
           75000, 70000, 80000, 85000, 90000, 95000]


categories = ["Electronics", "Clothing", "Furniture", "Grocery", "Sports"]
category_sales = [80000, 60000, 45000, 70000, 40000]


age = [22, 25, 28, 30, 32, 35, 38, 40, 42, 45,
       23, 27, 29, 31, 34, 36, 39, 41, 44, 46,
       24, 26, 33, 37, 43, 47, 50, 52, 55, 58]


discount = [5, 10, 15, 20, 25, 30, 35, 40, 45, 50]
revenue_discount = [90000, 85000, 82000, 78000, 75000,
                    70000, 65000, 62000, 58000, 55000]

# -----------------------------
# Create Dashboard
# -----------------------------

plt.figure(figsize=(14, 10))

# 1. Monthly Revenue Line Chart
plt.subplot(2, 2, 1)
plt.plot(months, revenue, marker="o", label="Revenue")
plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue (₹)")
plt.legend()
plt.grid(True)

# 2. Category-wise Sales Bar Chart
plt.subplot(2, 2, 2)
plt.bar(categories, category_sales, label="Sales")
plt.title("Category-wise Sales")
plt.xlabel("Category")
plt.ylabel("Sales (₹)")
plt.legend()
plt.grid(axis="y")

# 3. Customer Age Histogram
plt.subplot(2, 2, 3)
plt.hist(age, bins=8, label="Customers")
plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Customers")
plt.legend()
plt.grid(True)

# 4. Revenue vs Discount Scatter Plot
plt.subplot(2, 2, 4)
plt.scatter(discount, revenue_discount, label="Customers")
plt.title("Revenue vs Discount")
plt.xlabel("Discount (%)")
plt.ylabel("Revenue (₹)")
plt.legend()
plt.grid(True)

# Overall dashboard title
plt.suptitle("Mini Business Analytics Dashboard", fontsize=16)

# Adjust spacing
plt.tight_layout(rect=[0, 0, 1, 0.96])

plt.show()
```
