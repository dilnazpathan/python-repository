```python
# question 24 Take a student's marks in 5 subjects. Calculate percentage and print grade
#according to these rules:
#• 90 or above → A+
#• 80–89 → A
#• 70–79 → B
# 60–69 → C
#• Below 60 → Fail Also print "Passed with Distinction" if percentage >= 75.

m1 = int(input("enter marks of subject1:"))
m2 = int(input("enter marks of subject2:"))
m3 = int(input("enter marks of subject3:"))
m4 = int(input("enter marks of subject4:"))
m5 = int(input("enter marks of subject5:"))
total = m1+m2+m3+m4+m5
percentage = total / 5
print(percentage)
if percentage >= 90:
    print("grade: A+")
elif percentage >= 80-89:
    print("grade:  A")
elif percentage >= 70-79:
    print("grade: B")
elif percentage >= 60-69:
    print("grade: C")
    
```
