```python
# question 10 Take a year and check if it is a leap year (div by 4 but not 100, or div by 400).

year = int(input("enter a year:"))
if year % 4 == 0 and year % 100 != 0 or year % 400 == 0:
    print("leap year")
else:
    print("not a leap year")
```

```python
# question 15 Take three sides of a triangle. Check if they form a valid triangle. If yes, also print if it is Equilateral, Isosceles, or Scalene.

a = int(input("enter first side:"))
b = int(input("enter second side:"))
c = int(input("enter third side:"))
if a == b == c:
    print("print equilateral")
elif a == b and b == c and c == a:
    print("isosceles")
else:
    print("scalene")
```

```python
# question 16 Take student marks in 5 subjects. Calculate percentage and give grade: 90+ → A+, 80-89 → A, 70-79 → B, 60-69 → C, below 60 → Fail.

m1 = int(input("enter marks of first subject:"))
m2 = int(input("enter marks of seond subject:"))
m3 = int(input("enter marks of third subject:"))
m4 = int(input("enter marks of fourth subject:"))
m5 = int(input("enter marks of fifth subject:"))
total = m1+m2+m3+m4+m5
percentage =total / 5
if percentage >= 90:
    print("A+")
elif percentage >= 80-90:
    print("A")
elif percentage >= 70-79:
    print("B")
elif percentage >= 60-69:
    print("C")
else:
    print("fail")
```
