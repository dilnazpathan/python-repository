### Hard level

```python
# question 16 Take a sentence. Split into words, remove all words that start with vowel (a,e,i,o,u), then join remaining words.

sentence = input("Enter the sentence:")
words = sentence.split()
result = " ".join(filter(lambda x: x [0].lower() not in "aeiou",words))
print(result)
```

```python
# question 17 Take a string. Replace all spaces with -, then split by -, and print only words with length > 4.
s = input("Enter a string:")
s = s.replace(" ","-")
words = s.split("-")
for word in words:
    if len(word) > 4:
        print(word)
        

        
```

```python
# question 18 Create a list of student names. Join them with ", ". Then replace the first name with "New Student".

list = ["john","ram","shyam","seeta","geeta"]
print(",".join(list))
list[0] = "New student"
print(list)

```

```python
# question 19 Take a paragraph. Split it into sentences (using .), strip each sentence, and join them with "\n- ".

paragraph = input("Enter paragraph:")
sentences = paragraph.split(",")
result = []
for s in sentences:
    s = s.strip()
    if s:
        result.append(s)
print("-" + "\n-".join(result))
```

```python
# question 20 Take a string. Use replace(), strip(), split(), and join() together to clean and reformat it as a proper sentence.

s = input("Enter a string:")
s = s.replace("_"," ")
s = s.strip()
words = s.split()
result = " ".join(words)
print(result.capitalize())
```
