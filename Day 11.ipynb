```python
# question 24 Take two numbers start and end from the user. Using while loop, print all Armstrong numbers between start and end.
# (Example: 153 is Armstrong because 13 + 53 + 33 = 153)

start = int(input("enter start:"))
end = int(input("enter end:"))
num = start
while num <= end:
    temp = num
    digits = len(str(num))
    total = 0
    while temp > 0:
        digit = temp % 10
        total += digit ** digits
        temp //= 10
    if total == num:
        print(num)
    num += 1
    
```

```python
# question 25 Create a number guessing game using while loop:
#• Generate a random number between 1 to 100.
#• Give the user 7 chances to guess.
#• After each guess, tell whether the number is "Too High" or "Too Low".
#• If user guesses correctly, print "Congratulations!" and stop.
#• If chances are over, reveal the number.

import random 
secret = random.randint(1,100)
chance = 1
while chance <= 7:
    guess = int(input("enter your guess:"))
    if guess == secret:
        print("congratulations! you guessedcorrectly.")
        break
    elif guess < secret:
        print("too low")
    else:
        print("too high")
    chance += 1
if chance > 7:
    print("game over")
    print("the correct number was:",secret)
```

```python
# question 6 Write a program to check whether a number is palindrome or not.
# Palindrome means same from both sides.
# Example:
#121 → Palindrome
#1331 → Palindrome
#Steps:
#• Reverse the number
#• Compare with original number

num = int(input("enter number:"))
temp = num
reverse = 0
while temp > 0:
    digit = temp % 10
    reverse = reverse * 10 + digit
    temp = temp // 10
if num == reverse:
    print("palindrome")
else:
    print("not palindroe")
```

```python
# question 16 Check leap year repeatedly until user types "stop".
#Explanation
#• Ask year from user
#• Check leap year condition
#• Continue until "stop" entered
#Leap year condition:
#• Divisible by 4
#• Not divisible by 100 OR divisible by 400

while True:
    year = input("enter year(or type 'stop' to exit):")
    if year.lower() == "stop":
        break
    year = int(year)
    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
        print(year, "is leap year")
    else:
        print(year, "is not a leap year")
```

```python
# question 19 Keep taking student marks until negative number entered.
#Explanation
#• Add all marks
#• Count total students
#• Stop on negative input
#• Find average:
#average = total / count

total_marks = 0
student_count = 0
while True:
    marks = float(input("enter student marks(or any negative number to stop):"))
    if marks < 0:
        break
    total_marks += marks
    student_count >= 1
if student_count > 0:
    average = total_marks/student_count
    print("result")
    print(f"total students: {student_count}")
    print(f"average marks:{average:2f}")
else:
    print("valid marks were entered")
```

```python
# question 20 Create login system with maximum 3 attempts.
#Explanation
#• Store correct username and password
#• Ask user input
#• If correct → Login successful
#• If wrong → decrease attempts
#• After 3 wrong attempts → block account

correct_username = "admin"
correct_password = "password123"
attempts_left = 3
while attempts_left > 0:
    username = input("enter username:")
    password = input("enter password:")
    if username == correct_username and password == correct_password:
        print("login succeful!")
    else:
        attempts_left -= 1
        print(f"incorrect username or password. attempts left:{attempts_left}")
    if attempts_left == 0:
        print("your account has been blocked  after 3 failed attempts")
```
