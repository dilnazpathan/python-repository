```python
# question 1 Class Name: Shape
#Problem
#Create an abstract class named Shape.
#Explanation
#A shape can have different formulas for area and perimeter. Therefore, create two abstract #methods:
#• area()
#• perimeter()
#Create subclasses like Circle and Rectangle and implement both methods.

from abc import ABC,abstractmethod
import math
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
    @abstractmethod
    def perimeter(self):
        pass
class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius
    def area(self):
        return math.pi * self.radius **2
    def perimeter(self):
        return 2 * math.pi * self.radius
class Rectangle(Shape):
    def __init__(self,length,width):
        self.length = length
        self.width = width
    def area(self):
        return self.length * self.width
    def perimeter(self):
        return 2 *(self.length + self.width)
c = Circle(5)
print("circle area:", c.area())
print("rectangle perimeter:",c.perimeter())
r = Rectangle(10,4)
print("rectangle area:",r.area())
print("raectangle perimetr:",r.perimeter())
```

```python
# question 15 Class Name: Shape3D
#Problem
#Create an abstract class named Shape3D.
#Explanation
#Three-dimensional shapes have different formulas.
#Create abstract methods:
#• volume()
#• surface_area()
#Create subclasses:
#• Cube
#• Sphere

from abc import ABC,abstractmethod
import math
class Shape3D:
    @abstractmethod
    def volume(self):
        pass
    @abstractmethod
    def surface_area(self):
        pass
class Cube(Shape3D):
    def __init__(self,side):
        self.side  = side
    def volume(self):
        return self.side ** 3
    def surface_area(self):
        return 6 * self.side ** 2
class Sphere(Shape3D):
    def __init__(self,radius):
        self.radius = radius
    def volume(self):
        return (4/3) * math.pi * self.radius ** 3
    def surface_area(self):
        return 4 * math.pi * self.radius ** 2

cube = Cube(4)
print("cube volume:",cube.volume())
print("cube surface_area:",cube.surface_area())
sphere = Sphere(3)
print("sphere volume:",sphere.volume())
print("sphere surface_area:",sphere.surface_area())
```

```python
# question 32 Class Name: Shape
#Problem
#Create an abstract class named Shape.
#Explanation
#Every shape has a color, but different shapes calculate area and perimeter differently.
#Create a constructor:
#__init__(self, color)
#Store the color.
#Create two abstract methods:
#• area()
#• perimeter()
#Create one normal method:
#display_color()
#This method should display the color of the shape.
#Create the following subclasses
#Rectangle
#• Use super() to initialize the color.
#• Store length and width.
#• Implement area().
#• Implement perimeter().
#• Create a method to display all rectangle details.
#• Reuse the parent's display_color() method using super().
#Circle
#• Use super() to initialize the color.
#• Store the radius.
#• Implement area().
#• Implement perimeter().
#• Display all details.
#• Call the parent method using super().
#Concepts Covered
#• Abstract Class
#• Constructor
#• super()
#• Parent Method
#• Method Overriding
#• Code Reusability

from abc import ABC,abstractmethod
import math
class Shape(ABC):
    def __init__(self,color):
        self.color = color
    def display_color(self):
        print("color:",self.color)
    @abstractmethod
    def area(self):
        pass
    @abstractmethod
    def perimeter(self):
        pass
class Rectangle(Shape):
    def __init__(self,color,length,width):
        super().__init__(color)
        self.length = length
        self.width = width
    def area(self):
        return self.length * self.width
    def perimeter(self):
        return 2 *(self.length + self.width)
    def display_details(self):
        super().display_color()
        print("length:",self.length)
        print("width:",self.width)
        print("area:",self.area())
        print("perimeter:",self.perimeter())
class Circle(Shape):
    def __init__(self,color,radius):
        super().__init__(color)
        self.radius = radius
    def area(self):
        return math.pi * self.radius ** 2
    def perimeter(self):
        return 2 * math.pi * self.radius
    def display_details(self):
        super().display_color()
        print("radius:",self.radius)
        print("area:",round(self.area(),2))
        print("perimeter:",round(self.perimeter(),2))
r = Rectangle("blue",10,5)
print("reactangle details")
r.display_details()
c = Circle("red",7)
print("circle details")
c.display_details()
```

```python
# question 33 Class Name: Employee
#Problem
#Create an abstract class named Employee.
#Explanation
#Every employee has common information, but different employee roles receive different bonuses.
#Create a constructor:
#__init__(self, name, base_salary)
#Store:
#• Employee Name
#• Base Salary
#Create one normal method:
#display_employee()
#This method should display the employee information.
#Create one abstract method:
#calculate_bonus()
#Create the following subclasses
#Manager
#• Use super() in the constructor.
#• Bonus = 30% of base salary + additional bonus amount.
#• Override display_employee().
#• Call the parent version using super() before displaying manager-specific information.
#Developer
#• Use super() in the constructor.
#• Bonus = 20% of base salary.
#• Override display_employee().
#• Reuse the parent's method using super().
#Concepts Covered
#• Abstract Class
#• Constructor
#• super()
#• Parent Method
#• Method Overriding
#• Polymorphism

from abc import ABC,abstractmethod
class Employee(ABC):
    def __init__(self,name,base_salary):
        self.name = name
        self.base_salary = base_salary
    def display_employee(self):
        print("employee name:",self.name)
        print("base name:",self.base_salary)
    @abstractmethod
    def calculate_bonus(self):
        pass
class Manager(Employee):
    def __init__(self,name,base_salary,extra_bonus):
        super().__init__(name,base_salary)
        self.extra_bonus = extra_bonus
    def calculate_bonus(self):
        return (0.30 * self.base_salary) + self.extra_bonus
    def display_employee(self):
        super().display_employee()
        print("designation: manager")
        print("extra bonus:",self.extra_bonus)
        print("total nonus:",self.calculate_bonus())
class Developer(Employee):
    def __init__(self,name,base_salary):
        super().__init__(name,base_salary)
    def calculate_bonus(self):
        return 0.20 * self.base_salary
    def display_employee(self):
        super().display_employee()
        print("designation: developer")
        print("total bonus:",self.calculate_bonus())
m = Manager("dilnaz",50000,10000)
print("manager details")
m.display_employee()
print()
d = Developer("ruchika",40000)
print("developer details")
d.display_employee()
```

```python
# question 34 Class Name: Vehicle
#Problem
#Create an abstract class named Vehicle.
#Explanation
#Every vehicle has common information like brand and fuel type, but different vehicles start and consume fuel differently.
#Create a constructor:
#__init__(self, brand, fuel_type)
#Store:
#• Brand
#• Fuel Type
#Create a normal method:
#display_vehicle()
#Create abstract methods:
#• start()
#• fuel_consumption(distance)
#Create the following subclasses
#ElectricCar
#• Use super() to initialize parent attributes.
#• Validate that the fuel type is "Electric".
#• Implement start().
#• Implement fuel_consumption(distance) using battery consumption.
#• Override display_vehicle() and call the parent version using super().
#PetrolCar
#• Use super() to initialize parent attributes.
#• Validate that the fuel type is "Petrol".
#• Implement start().
#• Implement fuel_consumption(distance) using petrol consumption.
#• Override display_vehicle() and reuse the parent method using super().
#Concepts Covered
#• Abstract Class
#• Constructor
#• Validation
#• Method Overriding
#• super()
#• Parent Method

from abc import ABC, abstractmethod
class Vehicle(ABC):
    def __init__(self, brand, fuel_type):
        self.brand = brand
        self.fuel_type = fuel_type

    def display_vehicle(self):
        print("Brand:", self.brand)
        print("Fuel Type:", self.fuel_type)
    @abstractmethod
    def start(self):
        pass
    @abstractmethod
    def fuel_consumption(self, distance):
        pass
class ElectricCar(Vehicle):
    def __init__(self, brand, fuel_type):
        if fuel_type != "Electric":
            raise ValueError("Fuel type must be Electric")
        super().__init__(brand, fuel_type)
    def start(self):
        print(f"{self.brand} Electric Car started silently.")
    def fuel_consumption(self, distance):
        battery_used = distance * 0.15      # 0.15 units/km
        print("Battery Consumed:", battery_used, "units")
    def display_vehicle(self):
        super().display_vehicle()
        print("Vehicle Type: Electric Car")
class PetrolCar(Vehicle):
    def __init__(self, brand, fuel_type):
        if fuel_type != "Petrol":
            raise ValueError("Fuel type must be Petrol")
        super().__init__(brand, fuel_type)
    def start(self):
        print(f"{self.brand} Petrol Car started with engine.")
    def fuel_consumption(self, distance):
        petrol_used = distance * 0.08       # 0.08 litres/km
        print("Petrol Consumed:", petrol_used, "litres")
    def display_vehicle(self):
        super().display_vehicle()
        print("Vehicle Type: Petrol Car")
e1 = ElectricCar("Tesla", "Electric")
p1 = PetrolCar("Toyota", "Petrol")
print("Electric Car Details")
e1.display_vehicle()
e1.start()
e1.fuel_consumption(100)
print()
print("Petrol Car Details")
p1.display_vehicle()
p1.start()
p1.fuel_consumption(100)
```

```python
# question 35 Class Name: Report
#Problem
#Create an abstract class named Report.
#Explanation
#Every report has common information like title and data, but different report formats
#generate output differently.
#Create a constructor:
#__init__(self, title, data)
#Store:
#• Report Title
#• Report Data
#Create two normal methods:
#header()
#Displays the report title.
#footer()
#Displays a closing message indicating the report has ended.
#Create one abstract method:
#generate()
#Create the following subclasses
#PDFReport
#• Use super() in the constructor.
#• Inside generate():
#o Call super().header()
#o Generate the PDF report
#o Call super().footer()
#HTMLReport
#• Use super() in the constructor.
#• Inside generate():
#o Call super().header()
#o Generate the HTML report
#o Call super().footer()
#Concepts Covered
#• Abstract Class
#• Constructor
#• super()
#• Parent Methods
#• Method Overriding
#• Code Reusability
#• Real-world Report Generation

from abc import ABC, abstractmethod
class Report(ABC):
    def __init__(self, title, data):
        self.title = title
        self.data = data
    def header(self):
        print("REPORT ")
        print("Title:", self.title)
    def footer(self):
        print(" End of Report ")
    @abstractmethod
    def generate(self):
        pass
class PDFReport(Report):
    def __init__(self, title, data):
        super().__init__(title, data)
    def generate(self):
        super().header()
        print("Generating PDF Report")
        print("Report Data:", self.data)
        super().footer()
class HTMLReport(Report):
    def __init__(self, title, data):
        super().__init__(title, data)
    def generate(self):
        super().header()
        print("Generating HTML Report")
        print("Report Data:", self.data)
        super().footer()
pdf = PDFReport("Sales Report", "Sales increased by 20%")
html = HTMLReport("Student Report", "All students passed")
print("PDF Report")
pdf.generate()
print()
print("HTML Report")
html.generate()
```
