### Method Overloading Questions (1-7)

```python
# 1. Create a Calculator class that can add 2 numbers or 3 numbers using the same method name.
#Concept
#Method Overloading (Python uses default arguments or *args to achieve overloading.)
#What to do
#• Create a class Calculator.
#• Create one method add().
#• The method should work for either 2 numbers or 3 numbers.
#How to solve
#1. Create class Calculator.
#2. Define method add().
#3. Use default arguments or *args.
#4. Check how many numbers are provided.
#5. Return the sum.

class Calculator:
    def add(self,*numbers):
        if len(numbers) == 2:
            return numbers[0] + numbers[1]
        elif len(numbers) == 3:
            return numbers[0] + numbers[1] + numbers[2]
        else:
            return "enter only 2 or 3 numbers"
calc = Calculator()
print("sum of 2 numbers:", calc.add(10,20))
print("sum of 3 numbers:", calc.add(10,20,30))
```

```python
# 2. Create an Area class that calculates the area of a square and rectangle using the same method.
#Concept
#Method Overloading using default arguments.
#What to do
#• One parameter → Square.
#• Two parameters → Rectangle.
#How to solve
#1. Create class Area.
#2. Define method calculate().
#3. If only one value is provided, calculate the square's area.
#4. If two values are provided, calculate the rectangle's area.
#5. Return the result.

class Area:
    def calculate(self,length,width = None):
        if width is None:
            return length * length
        else:
            return length * width
area = Area()
print("area of square:",area.calculate(3))
print("area of reactange:",area.calculate(4,10))
```

```python
# 3. Create a Student class with a details() method that accepts different numbers of arguments.
#Concept
#Method Overloading
#What to do
#The method should work with:
#• Name
#• Name + Age
#• Name + Age + City
#How to solve
#1. Create class Student.
#2. Create method details().
#3. Use default arguments.
#4. Print the information based on the provided values.

class Student:
    def details(self,name,age = None,city = None):
        print("name:",name)
        if age is not None:
            
            print("age:",age)
        if city is not None:
            print("city:",city)
    print()
student = Student()
student.details("Rahul")
student.details("rahul",20)
student.details("rahul",20,"mumbai")
```

```python
# 4. Create a MathOperations class that multiplies any number of values.
#Concept
#Method Overloading using *args
#What to do
#Multiply:
#• 2 numbers
#• 3 numbers
#• 4 numbers
#• More numbers
#How to solve
#1. Create class MathOperations.
#2. Define method multiply(*args).
#3. Initialize the answer as 1.
#4. Loop through all numbers.
#5. Multiply each number.
#6. Return the final product.

class MathOperations:
    def multiply(self,*args):
        result = 1
        for num in args:
            result *= num
        return result
math = MathOperations()
print("product of 2 numbers:",math.multiply(2,3))
print("product of 3 numbers:",math.multiply(2,3,4))
print("product of 4 numbers:",math.multiply(2,3,4,5))
print("product of 5 numbers:",math.multiply(2,3,4,5,6))
```

```python
# 5. Create a Display class that prints different messages based on the argument type.
#Concept
#Method Overloading + Type Checking
#What to do
#Accept:
#• Integer
#• String
#• List
#How to solve
#1. Create class Display.
#2. Create method show().
#3. Use isinstance().
#4. Print different messages depending on the object's type.

class Display:
    def show(self,value):
        if isinstance(value,int):
            print("integer:",value)
        elif isinstance(value,list):
            print("list:",value)
        else:
            print("unsupported data type")
display = Display()
display.show(100)
display.show("hello pyhton")
display.show([10,20,30])
```

```python
# 6. Create a Volume class that calculates the volume of a cube and cuboid.
#Concept
#Method Overloading
#What to do
#• One parameter → Cube
#• Three parameters → Cuboid
#How to solve
#1. Create class Volume.
#2. Create method calculate().
#3. If one value is given, calculate the cube's volume.
#4. Otherwise, calculate the cuboid's volume.

class Volume:
    def calculate(self,length,width = None, height = None):
        if width is None and  height is None:
            return length ** 3
        else:
            return length * width * height
volume = Volume()
print("volume od cube:", volume.calculate(5))
print("volume of cuboid:",volume.calculate(5,4,3))
            
```

```python
# 7. Create a PrintData class that prints one or many values.
#Concept
#Method Overloading using *args
#What to do
#Print:
#• One value
#• Multiple values
#How to solve
#1. Create class PrintData.
#2. Create method print_data(*args).
#3. Loop through the arguments.
#4. Print each value.

class PrintData:
    def print_data(self,*args):
        for value in args:
            print(value)
data = PrintData()
print("one value:")
data.print_data(100)
print("\n multiple value:")
data.print_data(100,"hello python",3.12,True)
```

### Method Overriding Questions (8-14)

```python
# 8. Create an Animal class with sound(). Override it in Dog and Cat.
#Concept
#Method Overriding
#What to do
#• Animal → Generic sound
#• Dog → Bark
#• Cat → Meow
#How to solve
#1. Create parent class Animal.
#2. Define method sound().
#3. Inherit Dog and Cat.
#4. Override sound().
#5. Create objects and call the method.

class Animal:
    def sound(self):
        print("animal makes sound")
class Dog(Animal):
    def sound(self):
        print("dog says: barks")
class Cat(Animal):
    def sound(self):
        print("cat says: meow!")
dog = Dog()
cat = Cat()
dog.sound()
cat.sound()
```

```python
# 9. Create Vehicle, Car, and Bike classes.
#Concept
#Method Overriding
#What to do
#Override the start() method.
#How to solve
#1. Create Vehicle.
#2. Add method start().
#3. Create Car.
#4. Create Bike.
#5. Override start() in both child classes.


class Vehicle:
    def start(self):
        print("vehicle started")
class Car(Vehicle):
    def start(self):
        print("car started")
class Bike(Vehicle):
    def start(self):
        print("bike started")

c = Car()
B = Bike()

c.start()
B.start()
```

```python
# 10. Create an Employee class and override calculate_salary() in Manager.
#Concept
#Method Overriding
#What to do
#• Employee → Normal salary
#• Manager → Salary + Bonus
#How to solve
#1. Create parent class.
#2. Define calculate_salary().
#3. Inherit Manager.
#4. Override the method.
#5. Print the manager's salary.

class Employee:
    def calculate_salary(self):
        salary = 30000
        print("employee salary:",salary)
class Manager(Employee):
    def calculate_salary(self):
        salary = 50000
        bonus = 10000
        total_salary = salary + bonus
        print("manager salary:",total_salary)
employee = Employee()
manager = Manager()
employee.calculate_salary()
manager.calculate_salary()
    
```

```python
# 11. Create Shape, Circle, and Rectangle classes.
#Concept
#Method Overriding
#What to do
#Override the area() method.
#How to solve
#1. Create Shape.
#2. Add area().
#3. Inherit Circle.
#4. Inherit Rectangle.
#5. Override area() in each child class.

class Shape:
    def area(self):
        print("area of shape")
class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius
    def area(self):
        print("area of circle:",3.14 * self.radius * self.radius)
class Rectangle(Shape):
    def __init__(self,length,width):
        self.length = length
        self.width = width
    def area(self):
        print("area of rectangle:",self.length * self.width)
circle = Circle(7)
rectangle = Rectangle(10,5)
circle.area()
```

### Operator Overloading Questions (15-20)

```python
# 15. Create a Point class and overload the + operator.
#Concept
#Operator Overloading (__add__)
#What to do
#Add two points together.
#How to solve
#1. Create class Point.
#2. Store x and y coordinates.
#3. Overload __add__().
#4. Return a new point with added coordinates.

class Point:
    def __init__(self,x,y):
        self.x = x
        self.y = y
    def __add__(self,other):
        return Point(self.x + other.x, self.y + other.y)
    def display(self):
        print("point(",self.x,",",self.y,")",sep="")
p1 = Point(2,3)
p2 = Point(4,5)
p3 = p1+p2
print("first point:")
p1.display()
print("second point:")
p2.display()
print("result after addition:")
p3.display()
```

```python
# 16. Create a Book class and overload the + operator.
#Concept
#Operator Overloading
#What to do
#Add the number of pages of two books.
#How to solve
#1. Create class Book.
#2. Store pages.
#3. Overload __add__().
#4. Return the total pages.

class Book:
    def __init__(self,pages):
        self.pages = pages
    def __add__(self,other):
        return self.pages + other.pages
book1 = Book(150)
book2 = Book(250)
total_pages  = book1 + book2
print("pages in book 1:", book1.pages)
print("pages in book 2:",book2.pages)
print("total pages:",total_pages)
```

```python
# 17. Create a Student class and overload the > operator.
#Concept
#Comparison Operator Overloading (__gt__)
#What to do
#Compare marks of two students.
#How to solve
#1. Create class Student.
#2. Store marks.
#3. Overload __gt__().
#4. Compare the marks.
#5. Return True or False.

class Student:
    def __init__(self,marks):
        self.marks = marks
    def __gt__(self,other):
        return self.marks > other.marks
student1 = Student(85)
student2 = Student(75)
print("student1 marks:",student1.marks)
print("student2 marks:",student2.marks)
print("is student 1 greater than student 2?",student1 > student2)
```

```python
# 18. Create a Rectangle class and overload the == operator.
#Concept
#Equality Operator Overloading (__eq__)
#What to do
#Compare rectangles using their area.
#How to solve
#1. Create class Rectangle.
#2. Store length and width.
#3. Calculate area.
#4. Overload __eq__().
#5. Compare the areas.

class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width
    def __eq__(self, other):
        area1 = self.length * self.width
        area2 = other.length * other.width
        return area1 == area2
rect1 = Rectangle(5, 4)
rect2 = Rectangle(2, 10)
print("Rectangle 1 Area:", rect1.length * rect1.width)
print("Rectangle 2 Area:", rect2.length * rect2.width)
print("Are both rectangles equal?", rect1 == rect2)
```

```python
# 19. Create a Money class and overload the + and - operators.
#Concept
#Multiple Operator Overloading
#What to do
#Perform addition and subtraction of money amounts.
#How to solve
#1. Create class Money.
#2. Store the amount.
#3. Overload __add__().
#4. Overload __sub__().
#5. Return new Money objects after each operation.


class Money:
    def __init__(self, amount):
        self.amount = amount
    def __add__(self, other):
        return Money(self.amount + other.amount)
    def __sub__(self, other):
        return Money(self.amount - other.amount)
    def display(self):
        print("Amount:", self.amount)
money1 = Money(1000)
money2 = Money(400)
add_result = money1 + money2
sub_result = money1 - money2
print("Money 1:")
money1.display()
print("Money 2:")
money2.display()
print("After Addition:")
add_result.display()
print("After Subtraction:")
sub_result.display()
```

```python
# 20. Create a Distance class and overload <, >, and ==.
#Concept
#Comparison Operator Overloading
#What to do
#Compare two distance objects.
#How to solve
#1. Create class Distance.
#2. Store the distance value.
#3. Overload:
#o __lt__() for <
#o __gt__() for >
#o __eq__() for ==
#4. Compare the stored distance values.
#5. Return True or False.

class Distance:
    def __init__(self, value):
        self.value = value
    def __lt__(self, other):
        return self.value < other.value
    def __gt__(self, other):
        return self.value > other.value
    def __eq__(self,other):
        return self.value == other.value
d1 = Distance(15)
d2 = Distance(20)
print("distance 1:",d1.value)
print("distance 2:",d1.value)
print("d1 < d2:",d1 < d2)
print("d1 > d2:",d1 > d2)
print("d1 == d2:",d1 == d2)
```
