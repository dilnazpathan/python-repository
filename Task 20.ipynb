### Single-Level Inheritance (1–15)

```python
 # question 1 Class Name: Person → Student
#Problem
#Create a parent class named Person and a child class named Student.
#Explanation
#The Person class should store:
#• name
#• age
#The Student class should inherit from Person and add:
#• marks
#Create a method in Student to display the student's name, age, and marks.

from abc import ABC,abstractmethod
class Person(ABC):
    def __init__(self,name,age):
        self.name = name
        self.age = age
class Student(Person):
    def __init__(self,name,age,marks):
        super().__init__(name,age)
        self.marks = marks
    def display(self):
        print("name:",self.name)
        print("age:",self.age)
        print("marks:",self.marks)
s = Student("dilnaz",23,90)
s.display()
        
```

```python
# question 5 Class Name: Shape → Circle
#Problem
#Create a parent class named Shape.
#Explanation
#Create a method:
#• area()
#Create child class:
#• Circle
#Store:
#• radius
#Override area() to calculate the area of a circle.

class Shape:
    def area(self):
        print("area of the shape:")
class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius
    def area(self):
        area = 3.14 * self.radius * self.radius
        print("area of circle:",area)
c = Circle(5)
c.area()
```

```python
# question 7 Class Name: Product → Electronics
#Problem
#Create a parent class named Product.
#Explanation
#Store:
#• price
#Create method:
#• discount()
#Create child class:
#• Electronics
#Add:
#• warranty
#Display all information.

class Product:
    def __init__(self,price):
        self.price = price
    def discount(self,percent):
        discount = (self.price * percent) / 100
        final_price = self.price - discount
        print("after discount:",final_price)
class Electronics(Product):
    def __init__(self,price,warranty):
        super().__init__(price)
        self.warranty = warranty
    def display(self):
        print("price:",self.price)
        print("warranty:",self.warranty)
e = Electronics(50000,"2 years")
e.display()
e.discount(10)
        
```

```python
# question 26 Class Name:
#Account → SavingAccount → SeniorCitizenAccount
#Problem
#Create multi-level inheritance.
#Explanation
#Each level should implement different interest rules.

class Account:
    def __init__(self,balance):
        self.balance = balance
    def calculate_interest(self):
        interest = self.balance * 4/100
        print("account interest:",interest)
class SavingAccount(Account):
    def calculate_interest(self):
        interest = self.balance * 6/100
        print("saving account interest:",interest)
class SeniorCitizenAccount(SavingAccount):
    def calculate_interest(self):
        interest = self.balance * 8/100
        print("senior citizen interest:",interest)
s = SeniorCitizenAccount(100000)
s.calculate_interest()
```

```python
# question 29 Class Name:
#Shape → TwoDShape → Triangle
#Problem
#Create multi-level inheritance.
#Explanation
#Triangle should calculate its area using Heron's Formula.
#Store:
#• side1
#• side2
#• side3
#Implement the area calculation.
import math
class Shape:
    pass
class TwoDShape(Shape):
    pass
class Triangle(TwoDShape):
    def __init__(self,side1,side2,side3):
        self.side1 = side1
        self.side2 = side2
        self.side3 = side3
    def area(self):
        s = (self.side1+self.side2+self.side3)/2
        area = math.sqrt(s*(s-self.side1)*(s-self.side2)*(s-self.side3))
        print("area of triangle:",area)
       
t = Triangle(3,4,5)
t.area()
```

```python
# question 30 Class Name:
#GrandParent → Parent → Child → GrandChild
#Problem
#Create four levels of inheritance.
#Explanation
#Each class should have its own method.
#The GrandChild object should be able to access methods from all parent classes.
#Demonstrate Python's Method Resolution Order (MRO) using:
#ClassName.mro()
#or
#help(ClassName)

class GrandParent:
    def grandparent_method(self):
        print("Grandparent method")
class Parent(GrandParent):
    def parent_method(self):
        print("parent method")
class Child(Parent):
    def child_method(self):
        print("child method")
class GrandChild(Child):
    def GrandChild_method(self):
        print("GrandChild method")
g = GrandChild()
g.grandparent_method()
g.GrandChild_method()
g.child_method()
g.parent_method()
print(GrandChild.mro())   
```
