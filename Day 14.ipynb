```python
# question 32 Create a function make_multiplier(n) that returns another function which
#multiplies any number by n. (Use closure concept). Then create double = make_multiplier(2)
#and triple = make_multiplier(3) and test them.

def make_multiplier(n):
    def multiply(x):
        return x * n
    return multiply
double = make_multiplier(2)
triple = make_multiplier(3)
print(double(10))
print(triple(10))
```

```python
# question 33 Write a function smart_divide(func) as a decorator that:
#• Takes a function as argument
#• Checks if the second argument is zero before division
#• If zero, prints "Cannot divide by zero!" and returns None
#• Otherwise, calls the original function

def smart_divide(func):
    def wrapper(a,b):
        if b == 0:
            print("Cannot divide by zero!")
            return None
        return func(a,b)
    return wrapper
@smart_divide
def divide(a,b):
    return a/b
print(divide(10,2))
print(divide(10,0))
```

```python
# question 34 Create a function flexible_calc(*args, operation="add", **kwargs) that:
#• Can take any number of numbers
#• Supports operations: add, subtract, multiply, divide
#• If round_result=True in kwargs, round the final answer to 2 decimal places
#• Handle all edge cases

def flexible_calc(*args,operation="add",**kwargs):
    if len (args) == 0:
        return None
    if operation == "add":
        result = sum(args)
    elif operation == "substract":
        result = args[0]
        for i in args[1:]:
            result -= i
    elif operation == "multiply":
        result = 1
        for i in args:
            result *= i
    elif operation == "divide":
        result = args[0]
        for i in args[1:]:
            if i == 0:
                return "cannot divide by zero"
                result /= i
    else:
        return "invalid operation"
    if kwargs.get("round_result"):
        result = round(result,2)
    return result
print(flexible_calc(10,20,30))
print(flexible_calc(10,5,operation = "substract"))
print(flexible_calc(2,3,4, operation = "multiply"))
print(flexible_calc(20,3,operation = "divide", round_result = True))
    
    
```

```python
# question 35 Write a recursive function flatten_list(nested_list) that takes a nested list (list inside
#list) and returns a single flat list. Example: flatten_list([1, [2, 3], [4, [5, 6]]]) → [1, 2, 3, 4, 5, 6]

def flatten_list(nested_list):
    result = []
    for item in nested_list:
        if isinstance(item,list):
            result.extend(flatten_list(item))
        else:
            result.append(item)
    return result
print(flatten_list([1,[2,3],[4,[5,6]]]))
```
